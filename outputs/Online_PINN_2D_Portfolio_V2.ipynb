````python
# Cell 10B-V - 2D Wave exact solution autodiff 및 BC/IC 사전 검증

from datetime import datetime
import json


WAVE2D_EXACT_VERIFICATION_DTYPE = tf.float64
wave2d_exact_verification_tolerances = {
    "residual_max_abs": 1e-10,
    "boundary_exact_max_abs": 1e-10,
    "boundary_label_max_abs": 1e-6,
    "initial_displacement_label_max_abs": 1e-6,
    "initial_velocity_max_abs": 1e-10,
}


def wave2d_exact_solution_tf(coordinates):
    coordinates = tf.convert_to_tensor(
        coordinates,
        dtype=WAVE2D_EXACT_VERIFICATION_DTYPE,
    )
    x_value = coordinates[:, 0:1]
    y_value = coordinates[:, 1:2]
    t_value = coordinates[:, 2:3]
    dtype = coordinates.dtype
    pi = tf.constant(np.pi, dtype=dtype)
    kx = tf.constant(wave2d_eq["kx"], dtype=dtype)
    ky = tf.constant(wave2d_eq["ky"], dtype=dtype)
    wave_speed = tf.constant(wave2d_eq["wave_speed"], dtype=dtype)
    omega = wave_speed * pi * tf.sqrt(kx ** 2 + ky ** 2)
    return (
        tf.sin(pi * kx * x_value)
        * tf.sin(pi * ky * y_value)
        * tf.cos(omega * t_value)
    )


def wave2d_exact_derivatives_tf(coordinates):
    coordinates = tf.convert_to_tensor(
        coordinates,
        dtype=WAVE2D_EXACT_VERIFICATION_DTYPE,
    )
    x_value = coordinates[:, 0:1]
    y_value = coordinates[:, 1:2]
    t_value = coordinates[:, 2:3]
    with tf.GradientTape(
        persistent=True,
        watch_accessed_variables=False,
    ) as tape2:
        tape2.watch([x_value, y_value, t_value])
        with tf.GradientTape(
            persistent=True,
            watch_accessed_variables=False,
        ) as tape1:
            tape1.watch([x_value, y_value, t_value])
            values = wave2d_exact_solution_tf(
                tf.concat([x_value, y_value, t_value], axis=1)
            )
        u_x = tape1.gradient(values, x_value)
        u_y = tape1.gradient(values, y_value)
        u_t = tape1.gradient(values, t_value)
    u_xx = tape2.gradient(u_x, x_value)
    u_yy = tape2.gradient(u_y, y_value)
    u_tt = tape2.gradient(u_t, t_value)
    del tape1
    del tape2
    derivatives = {
        "u": values,
        "u_x": u_x,
        "u_y": u_y,
        "u_t": u_t,
        "u_xx": u_xx,
        "u_yy": u_yy,
        "u_tt": u_tt,
    }
    missing = [name for name, value in derivatives.items() if value is None]
    if missing:
        raise RuntimeError(f"Exact derivative computation failed: {missing}")
    return derivatives


def wave2d_exact_residual_tf(coordinates, lambda_value):
    derivatives = wave2d_exact_derivatives_tf(coordinates)
    dtype = derivatives["u"].dtype
    wave_speed_squared = tf.constant(
        wave2d_eq["wave_speed"] ** 2,
        dtype=dtype,
    )
    coefficient = tf.constant(lambda_value, dtype=dtype)
    return (
        derivatives["u_tt"]
        - coefficient
        * wave_speed_squared
        * (derivatives["u_xx"] + derivatives["u_yy"])
    )


wave2d_verification_coordinates = sample_lhs_2d(
    count=2048,
    seed=SEED + 2099,
    lower=[
        wave2d_domain["x_min"],
        wave2d_domain["y_min"],
        wave2d_domain["t_min"],
    ],
    upper=[
        wave2d_domain["x_max"],
        wave2d_domain["y_max"],
        wave2d_domain["t_max"],
    ],
).astype(np.float64)

exact_residual = wave2d_exact_residual_tf(
    wave2d_verification_coordinates,
    lambda_value=wave2d_eq["lambda_target"],
).numpy()
exact_residual_absolute = np.abs(exact_residual)

boundary_coordinates64 = tf.convert_to_tensor(
    X2_boundary.astype(np.float64),
    dtype=WAVE2D_EXACT_VERIFICATION_DTYPE,
)
boundary_exact = wave2d_exact_solution_tf(boundary_coordinates64).numpy()
boundary_label_error = boundary_exact - u2_boundary.astype(np.float64)

initial_coordinates64 = tf.convert_to_tensor(
    X2_initial.astype(np.float64),
    dtype=WAVE2D_EXACT_VERIFICATION_DTYPE,
)
with tf.GradientTape(watch_accessed_variables=False) as initial_tape:
    initial_tape.watch(initial_coordinates64)
    initial_exact = wave2d_exact_solution_tf(initial_coordinates64)
initial_gradient = initial_tape.gradient(initial_exact, initial_coordinates64)
if initial_gradient is None:
    raise RuntimeError("Exact initial derivative computation failed")
initial_velocity_exact = initial_gradient[:, 2:3].numpy()
initial_displacement_error = (
    initial_exact.numpy() - u2_initial.astype(np.float64)
)

wave2d_exact_verification = {
    "status": "pending",
    "problem": "wave2d",
    "config_fingerprint": wave2d_config_fingerprint,
    "dtype": "float64",
    "verification_points": int(len(wave2d_verification_coordinates)),
    "lambda_target": float(wave2d_eq["lambda_target"]),
    "exact_residual": {
        "mean_abs": float(np.mean(exact_residual_absolute)),
        "max_abs": float(np.max(exact_residual_absolute)),
        "rms": float(np.sqrt(np.mean(np.square(exact_residual)))),
    },
    "constraints": {
        "boundary_exact_max_abs": float(np.max(np.abs(boundary_exact))),
        "boundary_label_max_abs": float(
            np.max(np.abs(boundary_label_error))
        ),
        "initial_displacement_label_max_abs": float(
            np.max(np.abs(initial_displacement_error))
        ),
        "initial_velocity_max_abs": float(
            np.max(np.abs(initial_velocity_exact))
        ),
    },
    "tolerances": wave2d_exact_verification_tolerances,
    "generated_at": datetime.now().astimezone().isoformat(timespec="seconds"),
}

verification_checks = {
    "residual": (
        wave2d_exact_verification["exact_residual"]["max_abs"]
        <= wave2d_exact_verification_tolerances["residual_max_abs"]
    ),
    "boundary_exact": (
        wave2d_exact_verification["constraints"]["boundary_exact_max_abs"]
        <= wave2d_exact_verification_tolerances["boundary_exact_max_abs"]
    ),
    "boundary_label": (
        wave2d_exact_verification["constraints"]["boundary_label_max_abs"]
        <= wave2d_exact_verification_tolerances["boundary_label_max_abs"]
    ),
    "initial_displacement": (
        wave2d_exact_verification["constraints"]
        ["initial_displacement_label_max_abs"]
        <= wave2d_exact_verification_tolerances[
            "initial_displacement_label_max_abs"
        ]
    ),
    "initial_velocity": (
        wave2d_exact_verification["constraints"]["initial_velocity_max_abs"]
        <= wave2d_exact_verification_tolerances["initial_velocity_max_abs"]
    ),
}
wave2d_exact_verification["checks"] = verification_checks
wave2d_exact_verification["passed"] = all(verification_checks.values())
wave2d_exact_verification["status"] = (
    "passed" if wave2d_exact_verification["passed"] else "failed"
)

wave2d_exact_verification_path = (
    PROJECT_DIR / "report_artifacts" / "wave2d_exact_verification.json"
)
wave2d_exact_verification_path.write_text(
    json.dumps(wave2d_exact_verification, ensure_ascii=False, indent=2),
    encoding="utf-8",
)

print("2D Wave exact verification")
print(
    "Residual mean/max : "
    f"{wave2d_exact_verification['exact_residual']['mean_abs']:.6e} / "
    f"{wave2d_exact_verification['exact_residual']['max_abs']:.6e}"
)
for check_name, passed in verification_checks.items():
    print(f"{check_name:<22}: {'PASS' if passed else 'FAIL'}")
print(f"Verification artifact: {wave2d_exact_verification_path}")

if not wave2d_exact_verification["passed"]:
    raise RuntimeError(
        "2D Wave exact solution verification failed; training is blocked."
    )
````

````python
# Cell 10B-T - PDE 공통 Relative L1/L2 convergence utility

import importlib
import sys

convergence_module_code = r"""

from __future__ import annotations

from dataclasses import dataclass, field
import math

import numpy as np


CONVERGENCE_SCHEMA_VERSION = 1


def threshold_key(threshold):
    value = float(threshold)
    if not math.isfinite(value) or not 0.0 < value < 1.0:
        raise ValueError("Each convergence threshold must be finite and between 0 and 1")
    return format(value, ".8g")


def normalized_field_errors(exact, prediction, epsilon=1e-12):
    exact_array = np.asarray(exact, dtype=np.float64).reshape(-1)
    prediction_array = np.asarray(prediction, dtype=np.float64).reshape(-1)
    if exact_array.shape != prediction_array.shape:
        raise ValueError("Exact and prediction arrays must have identical shapes")
    if exact_array.size == 0:
        raise ValueError("Field error evaluation requires at least one value")
    if not np.all(np.isfinite(exact_array)) or not np.all(np.isfinite(prediction_array)):
        raise ValueError("Field error evaluation requires finite values")
    l1_denominator = float(np.sum(np.abs(exact_array)))
    l2_denominator = float(np.linalg.norm(exact_array))
    if l1_denominator <= epsilon or l2_denominator <= epsilon:
        raise ValueError("Exact field norm is too small for normalized error evaluation")
    error = prediction_array - exact_array
    return {
        "relative_l1": float(np.sum(np.abs(error)) / l1_denominator),
        "relative_l2": float(np.linalg.norm(error) / l2_denominator),
    }


@dataclass
class ConvergenceTargetTracker:
    thresholds: tuple
    metric_names: tuple = ("relative_l1", "relative_l2")
    evaluation_interval: int | None = None
    observations: list = field(default_factory=list)
    _targets: dict = field(init=False, repr=False)
    _last_iteration: int = field(default=-1, init=False, repr=False)
    _last_elapsed: float = field(default=-1.0, init=False, repr=False)

    def __post_init__(self):
        values = tuple(float(value) for value in self.thresholds)
        if len(values) != len(set(values)):
            raise ValueError("Convergence thresholds must be unique")
        for value in values:
            threshold_key(value)
        if tuple(sorted(values, reverse=True)) != values:
            raise ValueError("Convergence thresholds must be ordered from loose to strict")
        names = tuple(str(name).strip() for name in self.metric_names)
        if not names or any(not name for name in names):
            raise ValueError("At least one non-empty metric name is required")
        if self.evaluation_interval is not None and int(self.evaluation_interval) <= 0:
            raise ValueError("evaluation_interval must be positive")
        self.thresholds = values
        self.metric_names = names
        self._targets = {threshold_key(value): None for value in values}

    def observe(self, iteration, elapsed_seconds, metrics):
        iteration = int(iteration)
        elapsed_seconds = float(elapsed_seconds)
        if iteration < self._last_iteration:
            raise ValueError("Iterations must be observed in nondecreasing order")
        if elapsed_seconds < self._last_elapsed:
            raise ValueError("Elapsed time must be observed in nondecreasing order")
        values = {}
        for name in self.metric_names:
            value = float(metrics[name])
            if not math.isfinite(value) or value < 0.0:
                raise ValueError(f"Metric {name} must be finite and nonnegative")
            values[name] = value
        observation = {
            "iteration": iteration,
            "elapsed_seconds": elapsed_seconds,
            **values,
        }
        self.observations.append(observation)
        self._last_iteration = iteration
        self._last_elapsed = elapsed_seconds

        newly_reached = []
        for threshold in self.thresholds:
            key = threshold_key(threshold)
            if self._targets[key] is not None:
                continue
            if all(values[name] < threshold for name in self.metric_names):
                self._targets[key] = dict(observation)
                newly_reached.append(key)
        return newly_reached

    def snapshot(self, iteration_budget=None):
        targets = {}
        for threshold in self.thresholds:
            key = threshold_key(threshold)
            hit = self._targets[key]
            targets[key] = {
                "threshold": threshold,
                "reached": hit is not None,
                "iteration": None if hit is None else int(hit["iteration"]),
                "elapsed_seconds": None if hit is None else float(hit["elapsed_seconds"]),
                "metrics": (
                    None
                    if hit is None
                    else {name: float(hit[name]) for name in self.metric_names}
                ),
            }
        return {
            "schema_version": CONVERGENCE_SCHEMA_VERSION,
            "available": bool(self.observations),
            "criterion": "all_metrics_strictly_below_threshold",
            "metric_names": list(self.metric_names),
            "evaluation_interval": self.evaluation_interval,
            "iteration_budget": None if iteration_budget is None else int(iteration_budget),
            "observation_count": len(self.observations),
            "targets": targets,
        }


def unavailable_convergence_record(thresholds, metric_names, reason):
    targets = {
        threshold_key(value): {
            "threshold": float(value),
            "reached": None,
            "iteration": None,
            "elapsed_seconds": None,
            "metrics": None,
        }
        for value in thresholds
    }
    return {
        "schema_version": CONVERGENCE_SCHEMA_VERSION,
        "available": False,
        "criterion": "all_metrics_strictly_below_threshold",
        "metric_names": list(metric_names),
        "evaluation_interval": None,
        "iteration_budget": None,
        "observation_count": 0,
        "targets": targets,
        "reason": str(reason),
    }


def summarize_grouped_targets(
    results,
    group_names,
    thresholds,
    expected_runs,
    group_key="method",
    convergence_key="convergence",
):
    summaries = []
    for group_name in group_names:
        group_rows = [row for row in results if row.get(group_key) == group_name]
        for threshold in thresholds:
            key = threshold_key(threshold)
            available = [
                row
                for row in group_rows
                if row.get(convergence_key, {}).get("available") is True
                and key in row[convergence_key].get("targets", {})
            ]
            reached = [
                row[convergence_key]["targets"][key]
                for row in available
                if row[convergence_key]["targets"][key].get("reached") is True
            ]
            iterations = [float(item["iteration"]) for item in reached]
            elapsed = [float(item["elapsed_seconds"]) for item in reached]
            summaries.append(
                {
                    "group": group_name,
                    "threshold": float(threshold),
                    "expected_runs": int(expected_runs),
                    "available_runs": len(available),
                    "reached_runs": len(reached),
                    "not_reached_runs": len(available) - len(reached),
                    "coverage": len(available) / max(int(expected_runs), 1),
                    "reach_rate_among_available": (
                        None if not available else len(reached) / len(available)
                    ),
                    "median_iteration_reached": (
                        None if not iterations else float(np.median(iterations))
                    ),
                    "median_elapsed_seconds_reached": (
                        None if not elapsed else float(np.median(elapsed))
                    ),
                }
            )
    return summaries
"""

convergence_module_path = PROJECT_DIR / "app" / "ml" / "convergence.py"
convergence_module_path.write_text(convergence_module_code, encoding="utf-8")
importlib.invalidate_caches()
if "app.ml.convergence" in sys.modules:
    convergence_module = importlib.reload(sys.modules["app.ml.convergence"])
else:
    convergence_module = importlib.import_module("app.ml.convergence")

ConvergenceTargetTracker = convergence_module.ConvergenceTargetTracker
normalized_field_errors = convergence_module.normalized_field_errors
summarize_grouped_targets = convergence_module.summarize_grouped_targets
threshold_key = convergence_module.threshold_key

_utility_metrics = normalized_field_errors(
    exact=np.array([1.0, -1.0]),
    prediction=np.array([0.9, -0.9]),
)
_utility_tracker = ConvergenceTargetTracker(
    thresholds=(0.25, 0.01),
    evaluation_interval=1000,
)
assert _utility_tracker.observe(0, 0.0, _utility_metrics) == ["0.25"]
assert _utility_tracker.observe(
    1000,
    1.0,
    {"relative_l1": 0.009, "relative_l2": 0.009},
) == ["0.01"]
print("PDE convergence utility: PASS")
````

````python
# Cell 10C - 2D Wave-PINN 모델, loss 및 학습

import csv
import json
import time


class WavePINN2D(tf.keras.Model):
    def __init__(self, model_config):
        super().__init__()
        model_cfg = model_config["model"]
        domain_cfg = model_config["domain"]
        equation_cfg = model_config["equation"]
        self.wave_speed_squared = tf.constant(
            equation_cfg["wave_speed"] ** 2,
            dtype=tf.float32,
        )
        self.lb = tf.constant(
            [
                domain_cfg["x_min"],
                domain_cfg["y_min"],
                domain_cfg["t_min"],
            ],
            dtype=tf.float32,
        )
        self.ub = tf.constant(
            [
                domain_cfg["x_max"],
                domain_cfg["y_max"],
                domain_cfg["t_max"],
            ],
            dtype=tf.float32,
        )
        self.hidden_layers = [
            tf.keras.layers.Dense(
                units,
                activation=model_cfg["activation"],
                kernel_initializer="glorot_normal",
                bias_initializer="zeros",
            )
            for units in model_cfg["layers"][1:-1]
        ]
        self.output_layer = tf.keras.layers.Dense(
            model_cfg["layers"][-1],
            activation=None,
            kernel_initializer="glorot_normal",
            bias_initializer="zeros",
        )
        self.lambda_1 = self.add_weight(
            name="lambda_1",
            shape=(),
            initializer=tf.keras.initializers.Constant(
                equation_cfg["lambda_initial"]
            ),
            trainable=True,
            dtype=tf.float32,
        )

    def call(self, X, training=False):
        X = tf.cast(X, tf.float32)
        hidden = 2.0 * (X - self.lb) / (self.ub - self.lb) - 1.0
        for layer in self.hidden_layers:
            hidden = layer(hidden, training=training)
        return self.output_layer(hidden, training=training)

    def pde_residual(self, X):
        X = tf.cast(X, tf.float32)
        x_value = X[:, 0:1]
        y_value = X[:, 1:2]
        t_value = X[:, 2:3]
        with tf.GradientTape(
            persistent=True,
            watch_accessed_variables=False,
        ) as tape2:
            tape2.watch([x_value, y_value, t_value])
            with tf.GradientTape(
                persistent=True,
                watch_accessed_variables=False,
            ) as tape1:
                tape1.watch([x_value, y_value, t_value])
                prediction = self(
                    tf.concat([x_value, y_value, t_value], axis=1),
                    training=True,
                )
            u_x = tape1.gradient(prediction, x_value)
            u_y = tape1.gradient(prediction, y_value)
            u_t = tape1.gradient(prediction, t_value)
        u_xx = tape2.gradient(u_x, x_value)
        u_yy = tape2.gradient(u_y, y_value)
        u_tt = tape2.gradient(u_t, t_value)
        del tape1
        del tape2
        return (
            u_tt
            - self.lambda_1
            * self.wave_speed_squared
            * (u_xx + u_yy)
        )


def tensor2d(array):
    return tf.convert_to_tensor(array, dtype=tf.float32)


train2d = {
    "X_initial": tensor2d(X2_initial_train),
    "u_initial": tensor2d(u2_initial_train),
    "ut_initial": tensor2d(ut2_initial_train),
    "X_boundary": tensor2d(X2_boundary_train),
    "u_boundary": tensor2d(u2_boundary_train),
    "X_collocation": tensor2d(X2_collocation_train),
    "X_supervised": tensor2d(X2_supervised_train),
    "u_supervised": tensor2d(u2_supervised_train),
}
val2d = {
    "X_initial": tensor2d(X2_initial_val),
    "u_initial": tensor2d(u2_initial_val),
    "ut_initial": tensor2d(ut2_initial_val),
    "X_boundary": tensor2d(X2_boundary_val),
    "u_boundary": tensor2d(u2_boundary_val),
    "X_collocation": tensor2d(X2_collocation_val),
    "X_supervised": tensor2d(X2_supervised_val),
    "u_supervised": tensor2d(u2_supervised_val),
}
test2d = {
    "X_initial": tensor2d(X2_initial_test),
    "u_initial": tensor2d(u2_initial_test),
    "ut_initial": tensor2d(ut2_initial_test),
    "X_boundary": tensor2d(X2_boundary_test),
    "u_boundary": tensor2d(u2_boundary_test),
    "X_collocation": tensor2d(X2_collocation_test),
    "X_supervised": tensor2d(X2_supervised_test),
    "u_supervised": tensor2d(u2_supervised_test),
}

wave2d_loss_cfg = wave2d_config["loss_weights"]
wave2d_weight_sum = sum(wave2d_loss_cfg.values())
wave2d_weights = {
    name: value / wave2d_weight_sum
    for name, value in wave2d_loss_cfg.items()
}


def calculate_losses_2d(model2d, data2d, training=False):
    initial_coordinates = tf.cast(data2d["X_initial"], tf.float32)
    with tf.GradientTape(watch_accessed_variables=False) as initial_tape:
        initial_tape.watch(initial_coordinates)
        initial_prediction = model2d(initial_coordinates, training=training)
    initial_gradient = initial_tape.gradient(
        initial_prediction,
        initial_coordinates,
    )
    initial_velocity_prediction = initial_gradient[:, 2:3]
    loss_initial_displacement = tf.reduce_mean(
        tf.square(data2d["u_initial"] - initial_prediction)
    )
    loss_initial_velocity = tf.reduce_mean(
        tf.square(data2d["ut_initial"] - initial_velocity_prediction)
    )
    loss_initial = 0.5 * (
        loss_initial_displacement + loss_initial_velocity
    )
    boundary_prediction = model2d(
        data2d["X_boundary"],
        training=training,
    )
    loss_boundary = tf.reduce_mean(
        tf.square(data2d["u_boundary"] - boundary_prediction)
    )
    residual = model2d.pde_residual(data2d["X_collocation"])
    loss_physics = tf.reduce_mean(tf.square(residual))
    supervised_prediction = model2d(
        data2d["X_supervised"],
        training=training,
    )
    loss_supervised = tf.reduce_mean(
        tf.square(data2d["u_supervised"] - supervised_prediction)
    )
    total = (
        wave2d_weights["initial"] * loss_initial
        + wave2d_weights["boundary"] * loss_boundary
        + wave2d_weights["physics"] * loss_physics
        + wave2d_weights["supervised"] * loss_supervised
    )
    return {
        "total": total,
        "initial": loss_initial,
        "initial_displacement": loss_initial_displacement,
        "initial_velocity": loss_initial_velocity,
        "boundary": loss_boundary,
        "physics": loss_physics,
        "supervised": loss_supervised,
    }


tf.keras.utils.set_random_seed(SEED + 2100)
model2d = WavePINN2D(wave2d_config)
model2d(tf.zeros((1, 3), dtype=tf.float32))
training2d_cfg = wave2d_config["training"]
schedule2d = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=training2d_cfg["learning_rate"],
    decay_steps=training2d_cfg["decay_steps"],
    decay_rate=training2d_cfg["decay_rate"],
    staircase=True,
)
optimizer2d = tf.keras.optimizers.Adam(learning_rate=schedule2d)


@tf.function
def train_step_2d():
    with tf.GradientTape() as tape:
        losses = calculate_losses_2d(model2d, train2d, training=True)
    gradients = tape.gradient(losses["total"], model2d.trainable_variables)
    optimizer2d.apply_gradients(zip(gradients, model2d.trainable_variables))
    return losses


@tf.function
def validation_step_2d():
    return calculate_losses_2d(model2d, val2d, training=False)


best2d_path = wave2d_artifact_dir / "best_target.weights.h5"
last2d_path = wave2d_artifact_dir / "last.weights.h5"
final2d_path = wave2d_artifact_dir / "final.weights.h5"
history2d_path = PROJECT_DIR / "training" / "wave2d_training_history.csv"
baseline_target_path = (
    PROJECT_DIR / "training" / "wave2d_convergence_targets.json"
)
wave2d_result_path = PROJECT_DIR / "report_artifacts" / "wave2d_result.json"

history2d = []
baseline_convergence = None
reuse_verified_wave2d_baseline = (
    REUSE_VERIFIED_WAVE2D_BASELINE
    and final2d_path.exists()
    and wave2d_result_path.exists()
)

if reuse_verified_wave2d_baseline:
    model2d.load_weights(final2d_path)
    stored_wave2d_result = json.loads(
        wave2d_result_path.read_text(encoding="utf-8")
    )
    best_validation2d = float(
        stored_wave2d_result["evaluation"]["best_validation_loss"]
    )
    stored_training_run = stored_wave2d_result.get("training_run", {})
    baseline_protocol_version = int(
        stored_training_run.get("protocol_version", 1)
    )
    baseline_stop_reason = "restored_existing_api_artifact"
    baseline_iterations_completed = stored_training_run.get(
        "iterations_completed",
        stored_wave2d_result.get("training", {}).get("iterations"),
    )
    if history2d_path.exists():
        with history2d_path.open("r", encoding="utf-8") as stream:
            history2d = list(csv.DictReader(stream))
    print("2D API baseline training: skipped")
    print(f"Loaded artifact         : {final2d_path}")
    print(f"Stored protocol version : {baseline_protocol_version}")
    print(f"Stored best validation  : {best_validation2d:.6e}")

else:
    baseline_protocol_version = 2
    baseline_tracker = ConvergenceTargetTracker(
        thresholds=tuple(training2d_cfg["report_thresholds"]),
        metric_names=tuple(training2d_cfg["stop_rule"]["metric_names"]),
        evaluation_interval=training2d_cfg["evaluation_interval"],
    )
    baseline_primary_key = threshold_key(
        training2d_cfg["stop_rule"]["threshold"]
    )
    best_validation2d = np.inf
    best_target_score2d = np.inf
    baseline_started_at = time.perf_counter()

    def write_baseline_history():
        if not history2d:
            return
        with history2d_path.open("w", newline="", encoding="utf-8") as stream:
            writer = csv.DictWriter(stream, fieldnames=list(history2d[0]))
            writer.writeheader()
            writer.writerows(history2d)

    def record_baseline_checkpoint(iteration, losses):
        validation2d = validation_step_2d()
        current_validation2d = float(validation2d["total"].numpy())
        prediction2d = model2d(
            tensor2d(X2_exact),
            training=False,
        ).numpy()
        field_metrics2d = normalized_field_errors(
            exact=u2_exact,
            prediction=prediction2d,
        )
        elapsed2d = time.perf_counter() - baseline_started_at
        newly_reached = baseline_tracker.observe(
            iteration=iteration,
            elapsed_seconds=elapsed2d,
            metrics=field_metrics2d,
        )
        target_score2d = max(
            field_metrics2d["relative_l1"],
            field_metrics2d["relative_l2"],
        )
        values = validation2d if losses is None else losses
        record2d = {
            "iteration": int(iteration),
            "elapsed_seconds": float(elapsed2d),
            "total_loss": float(values["total"].numpy()),
            "initial_loss": float(values["initial"].numpy()),
            "initial_displacement_loss": float(
                values["initial_displacement"].numpy()
            ),
            "initial_velocity_loss": float(values["initial_velocity"].numpy()),
            "boundary_loss": float(values["boundary"].numpy()),
            "physics_loss": float(values["physics"].numpy()),
            "supervised_loss": float(values["supervised"].numpy()),
            "validation_loss": current_validation2d,
            "full_grid_relative_l1": field_metrics2d["relative_l1"],
            "full_grid_relative_l2": field_metrics2d["relative_l2"],
            "target_score_max_relative_l1_l2": target_score2d,
            "lambda_1": float(model2d.lambda_1.numpy()),
        }
        history2d.append(record2d)
        write_baseline_history()
        snapshot = baseline_tracker.snapshot(
            iteration_budget=training2d_cfg["maximum_iterations"]
        )
        baseline_target_path.write_text(
            json.dumps(snapshot, ensure_ascii=False, indent=2),
            encoding="utf-8",
        )
        reached_text = "-" if not newly_reached else ",".join(newly_reached)
        print(
            f"2D baseline iter={iteration:6d} "
            f"rel_l1={record2d['full_grid_relative_l1']:.4e} "
            f"rel_l2={record2d['full_grid_relative_l2']:.4e} "
            f"time={elapsed2d:.1f}s targets={reached_text} "
            f"lambda_1={record2d['lambda_1']:.6f}"
        )
        return current_validation2d, target_score2d, snapshot

    current_validation2d, target_score2d, baseline_convergence = (
        record_baseline_checkpoint(iteration=0, losses=None)
    )
    best_validation2d = current_validation2d
    best_target_score2d = target_score2d
    model2d.save_weights(best2d_path)
    baseline_iterations_completed = 0
    baseline_target_reached = (
        baseline_convergence["targets"][baseline_primary_key]["reached"] is True
    )

    if not baseline_target_reached:
        for iteration in range(
            1,
            training2d_cfg["maximum_iterations"] + 1,
        ):
            losses2d = train_step_2d()
            baseline_iterations_completed = iteration
            if (
                iteration % training2d_cfg["evaluation_interval"] == 0
                or iteration == training2d_cfg["maximum_iterations"]
            ):
                current_validation2d, target_score2d, baseline_convergence = (
                    record_baseline_checkpoint(iteration, losses2d)
                )
                best_validation2d = min(
                    best_validation2d,
                    current_validation2d,
                )
                if target_score2d < best_target_score2d:
                    best_target_score2d = target_score2d
                    model2d.save_weights(best2d_path)
                baseline_target_reached = (
                    baseline_convergence["targets"][baseline_primary_key]
                    ["reached"] is True
                )
                if baseline_target_reached:
                    break

    model2d.save_weights(last2d_path)
    model2d.load_weights(best2d_path)
    model2d.save_weights(final2d_path)
    baseline_stop_reason = (
        "relative_l1_l2_target_reached"
        if baseline_target_reached
        else "maximum_iterations_reached"
    )
    print(f"2D baseline stop reason : {baseline_stop_reason}")
    print(f"2D completed iterations : {baseline_iterations_completed}")
    print(f"2D best validation      : {best_validation2d:.6e}")
    print(f"2D best target weights  : {best2d_path}")
    print(f"2D final weights        : {final2d_path}")
````

````python
# Cell 10F - V2 Relative L1/L2 목표기반 2D 방법 비교

import csv
import hashlib
import json
import time

import matplotlib.pyplot as plt


for required_name in (
    'ConvergenceTargetTracker',
    'normalized_field_errors',
    'summarize_grouped_targets',
    'threshold_key',
):
    if required_name not in globals():
        raise RuntimeError(
            'Run Cell 10B-T before the V2 method comparison.'
        )


method_comparison_cfg = wave2d_config["method_comparison"]
method_specs = {
    "supervised_only": {
        "supervised_only": True,
        "fourier": False,
        "self_adaptive": False,
        "curriculum": False,
    },
    "baseline": {
        "supervised_only": False,
        "fourier": False,
        "self_adaptive": False,
        "curriculum": False,
    },
    "fourier": {
        "supervised_only": False,
        "fourier": True,
        "self_adaptive": False,
        "curriculum": False,
    },
    "self_adaptive": {
        "supervised_only": False,
        "fourier": False,
        "self_adaptive": True,
        "curriculum": False,
    },
    "fourier_curriculum": {
        "supervised_only": False,
        "fourier": True,
        "self_adaptive": False,
        "curriculum": True,
    },
    "fourier_self_adaptive": {
        "supervised_only": False,
        "fourier": True,
        "self_adaptive": True,
        "curriculum": False,
    },
}

configured_methods = method_comparison_cfg["methods"]
configured_seeds = method_comparison_cfg["seeds"]
expected_experiment_count = len(configured_methods) * len(configured_seeds)
if set(configured_methods) != set(method_specs):
    raise ValueError("2D comparison method configuration is inconsistent")
if len(configured_seeds) != 4 or len(set(configured_seeds)) != 4:
    raise ValueError("Exactly four unique comparison seeds are required")
if method_comparison_cfg["protocol_version"] != 2:
    raise ValueError("V2 target-stopping protocol is required")
comparison_stop_cfg = method_comparison_cfg["stop_rule"]
comparison_thresholds = tuple(
    float(value) for value in method_comparison_cfg["report_thresholds"]
)
if comparison_stop_cfg["threshold"] not in comparison_thresholds:
    raise ValueError("The stopping threshold must be included in report_thresholds")

comparison_root = PROJECT_DIR / method_comparison_cfg["artifact_directory"]
initialization_root = (
    PROJECT_DIR / method_comparison_cfg["initialization"]["directory"]
)
comparison_root.mkdir(parents=True, exist_ok=True)
initialization_root.mkdir(parents=True, exist_ok=True)

comparison_json_path = (
    PROJECT_DIR / "report_artifacts" / "wave2d_method_comparison.json"
)
comparison_csv_path = (
    PROJECT_DIR / "report_artifacts" / "wave2d_method_comparison.csv"
)
comparison_report_path = (
    PROJECT_DIR / "report_artifacts" / "wave2d_method_comparison.md"
)
comparison_corpus_path = (
    PROJECT_DIR / "data" / "documents" / "wave2d_method_comparison.md"
)

fingerprint_payload = {
    "equation": wave2d_config["equation"],
    "domain": wave2d_config["domain"],
    "model": wave2d_config["model"],
    "data": wave2d_config["data"],
    "loss_weights": wave2d_config["loss_weights"],
    "training": wave2d_config["training"],
    "comparison": method_comparison_cfg,
}
comparison_fingerprint = hashlib.sha256(
    json.dumps(
        fingerprint_payload,
        sort_keys=True,
        ensure_ascii=False,
    ).encode("utf-8")
).hexdigest()


class WavePINN2DComparison(WavePINN2D):
    def __init__(self, model_config, use_fourier, fourier_seed):
        super().__init__(model_config)
        self.use_fourier = bool(use_fourier)
        fourier_cfg = model_config["method_comparison"]["fourier"]
        if self.use_fourier:
            generator = np.random.default_rng(fourier_seed)
            matrix = generator.normal(
                loc=0.0,
                scale=fourier_cfg["scale"],
                size=(3, fourier_cfg["feature_count"]),
            ).astype(np.float32)
            self.fourier_matrix = tf.constant(matrix, dtype=tf.float32)
        else:
            self.fourier_matrix = None

    def call(self, X, training=False):
        X = tf.cast(X, tf.float32)
        normalized = 2.0 * (X - self.lb) / (self.ub - self.lb) - 1.0
        if self.use_fourier:
            projection = 2.0 * np.pi * tf.matmul(
                normalized,
                self.fourier_matrix,
            )
            hidden = tf.concat(
                [normalized, tf.sin(projection), tf.cos(projection)],
                axis=1,
            )
        else:
            hidden = normalized
        for layer in self.hidden_layers:
            hidden = layer(hidden, training=training)
        return self.output_layer(hidden, training=training)


def _file_sha256(path):
    digest = hashlib.sha256()
    with path.open("rb") as stream:
        for block in iter(lambda: stream.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()


def _initialization_group(method_spec):
    return "fourier" if method_spec["fourier"] else "plain"


def _initialize_comparison_model(method_name, method_spec, seed):
    group = _initialization_group(method_spec)
    weight_path = initialization_root / f"seed_{seed}_{group}.weights.h5"
    manifest_path = initialization_root / f"seed_{seed}_{group}.json"

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)
    model = WavePINN2DComparison(
        wave2d_config,
        use_fourier=method_spec["fourier"],
        fourier_seed=seed,
    )
    model(tf.zeros((1, 3), dtype=tf.float32))

    initialization_source = "deterministic_created"
    if weight_path.exists():
        model.load_weights(weight_path)
        initialization_source = "existing_file"
        if manifest_path.exists():
            manifest = json.loads(manifest_path.read_text(encoding="utf-8"))
            previous_initialization_fingerprint = manifest.get(
                "config_fingerprint"
            )
            if manifest.get("seed") != seed or manifest.get("group") != group:
                raise RuntimeError(
                    f"Initialization metadata mismatch: {manifest_path}"
                )
            expected_checksum = manifest.get("sha256")
            if expected_checksum and expected_checksum != _file_sha256(weight_path):
                raise RuntimeError(
                    f"Initialization checksum mismatch: {weight_path}"
                )
    else:
        model.save_weights(weight_path)

    manifest = {
        "protocol_version": 2,
        "seed": seed,
        "group": group,
        "compatible_methods": [
            name
            for name, spec in method_specs.items()
            if _initialization_group(spec) == group
        ],
        "config_fingerprint": comparison_fingerprint,
        "parameter_count": int(model.count_params()),
        "weights": str(weight_path.relative_to(PROJECT_DIR)),
        "sha256": _file_sha256(weight_path),
        "source": initialization_source,
    }
    manifest_path.write_text(
        json.dumps(manifest, ensure_ascii=False, indent=2),
        encoding="utf-8",
    )

    # Every experiment loads the persisted initialization, including its creator.
    model.load_weights(weight_path)
    return model, {
        "group": group,
        "weights": str(weight_path.relative_to(PROJECT_DIR)),
        "manifest": str(manifest_path.relative_to(PROJECT_DIR)),
        "sha256": manifest["sha256"],
        "source": initialization_source,
    }


def _curriculum_weights(iteration):
    curriculum = method_comparison_cfg["curriculum"]
    progress = tf.clip_by_value(
        tf.cast(iteration, tf.float32) / curriculum["ramp_iterations"],
        0.0,
        1.0,
    )
    physics = curriculum["physics_start"] + progress * (
        curriculum["physics_end"] - curriculum["physics_start"]
    )
    constraint = curriculum["constraint_start"] + progress * (
        curriculum["constraint_end"] - curriculum["constraint_start"]
    )
    raw = {
        "initial": constraint,
        "boundary": constraint,
        "physics": physics,
        "supervised": tf.constant(
            curriculum["supervised_weight"],
            dtype=tf.float32,
        ),
    }
    total = tf.add_n(list(raw.values()))
    return {name: value / total for name, value in raw.items()}


def _fixed_weights():
    raw = {
        name: tf.constant(value, dtype=tf.float32)
        for name, value in wave2d_config["loss_weights"].items()
    }
    total = tf.add_n(list(raw.values()))
    return {name: value / total for name, value in raw.items()}


def _self_adaptive_physics_loss(residual):
    adaptive_cfg = method_comparison_cfg["self_adaptive"]
    epsilon = tf.constant(adaptive_cfg["epsilon"], dtype=tf.float32)
    squared = tf.square(residual)
    residual_scale = tf.sqrt(squared + epsilon)
    normalized = residual_scale / (
        tf.reduce_mean(residual_scale) + epsilon
    )
    clipped = tf.clip_by_value(
        normalized,
        adaptive_cfg["minimum_weight"],
        adaptive_cfg["maximum_weight"],
    )
    weights = clipped / (tf.reduce_mean(clipped) + epsilon)
    weights = tf.stop_gradient(weights)
    return (
        tf.reduce_mean(weights * squared),
        tf.reduce_min(weights),
        tf.reduce_max(weights),
    )


def calculate_method_losses(model, data, method_spec, iteration, training=False):
    initial_coordinates = tf.cast(data["X_initial"], tf.float32)
    with tf.GradientTape(watch_accessed_variables=False) as initial_tape:
        initial_tape.watch(initial_coordinates)
        initial_prediction = model(initial_coordinates, training=training)
    initial_gradient = initial_tape.gradient(
        initial_prediction,
        initial_coordinates,
    )
    initial_velocity_prediction = initial_gradient[:, 2:3]
    loss_initial_displacement = tf.reduce_mean(
        tf.square(data["u_initial"] - initial_prediction)
    )
    loss_initial_velocity = tf.reduce_mean(
        tf.square(data["ut_initial"] - initial_velocity_prediction)
    )
    loss_initial = 0.5 * (
        loss_initial_displacement + loss_initial_velocity
    )
    boundary_prediction = model(data["X_boundary"], training=training)
    loss_boundary = tf.reduce_mean(
        tf.square(data["u_boundary"] - boundary_prediction)
    )
    residual = model.pde_residual(data["X_collocation"])
    if method_spec["self_adaptive"]:
        loss_physics, adaptive_min, adaptive_max = (
            _self_adaptive_physics_loss(residual)
        )
    else:
        loss_physics = tf.reduce_mean(tf.square(residual))
        adaptive_min = tf.constant(1.0, dtype=tf.float32)
        adaptive_max = tf.constant(1.0, dtype=tf.float32)
    supervised_prediction = model(data["X_supervised"], training=training)
    loss_supervised = tf.reduce_mean(
        tf.square(data["u_supervised"] - supervised_prediction)
    )
    weights = (
        _curriculum_weights(iteration)
        if method_spec["curriculum"]
        else _fixed_weights()
    )
    total = (
        weights["initial"] * loss_initial
        + weights["boundary"] * loss_boundary
        + weights["physics"] * loss_physics
        + weights["supervised"] * loss_supervised
    )
    return {
        "total": total,
        "initial": loss_initial,
        "initial_displacement": loss_initial_displacement,
        "initial_velocity": loss_initial_velocity,
        "boundary": loss_boundary,
        "physics": loss_physics,
        "supervised": loss_supervised,
        "adaptive_weight_min": adaptive_min,
        "adaptive_weight_max": adaptive_max,
    }


def _supervised_validation_mse(model):
    prediction = model(val2d["X_supervised"], training=False)
    return float(
        tf.reduce_mean(
            tf.square(val2d["u_supervised"] - prediction)
        ).numpy()
    )


def _physics_loss_at_target_lambda(model, data):
    original_lambda = float(model.lambda_1.numpy())
    model.lambda_1.assign(float(wave2d_eq["lambda_target"]))
    try:
        residual = model.pde_residual(data["X_collocation"])
        return float(tf.reduce_mean(tf.square(residual)).numpy())
    finally:
        model.lambda_1.assign(original_lambda)


def _full_grid_metrics(model):
    prediction = model(tensor2d(X2_exact), training=False).numpy()
    metrics = normalized_field_errors(
        exact=u2_exact,
        prediction=prediction,
    )
    return prediction, metrics


def _target_score(metrics):
    return max(
        float(metrics["relative_l1"]),
        float(metrics["relative_l2"]),
    )


def evaluate_comparison_model_v2(
    model,
    method_name,
    method_spec,
    seed,
    best_supervised_validation_mse,
    elapsed,
    completed_iterations,
    stop_reason,
    convergence,
    artifacts,
    initialization,
):
    common_losses = calculate_losses_2d(model, test2d, training=False)
    prediction, field_metrics = _full_grid_metrics(model)
    error = u2_exact - prediction
    lambda_trained = not method_spec["supervised_only"]
    lambda_value = float(model.lambda_1.numpy()) if lambda_trained else None
    lambda_error = (
        abs(float(wave2d_eq["lambda_target"]) - lambda_value)
        / float(wave2d_eq["lambda_target"])
        * 100.0
        if lambda_trained
        else None
    )
    primary_key = threshold_key(comparison_stop_cfg["threshold"])
    primary_target = convergence["targets"][primary_key]
    return {
        "protocol_version": 2,
        "experiment_family": "relative_l1_l2_target_stopping",
        "method": method_name,
        "seed": int(seed),
        "maximum_iterations": int(method_comparison_cfg["maximum_iterations"]),
        "iterations_completed": int(completed_iterations),
        "stop_reason": stop_reason,
        "target_reached": primary_target["reached"],
        "checkpoint_criterion": "minimum_max_relative_l1_l2",
        "best_supervised_validation_mse": float(
            best_supervised_validation_mse
        ),
        "full_grid_mse": float(np.mean(np.square(error))),
        "full_grid_mae": float(np.mean(np.abs(error))),
        "full_grid_relative_l1": field_metrics["relative_l1"],
        "full_grid_relative_l2": field_metrics["relative_l2"],
        "test_physics_loss_learned_lambda": float(
            common_losses["physics"].numpy()
        ),
        "test_physics_loss_target_lambda": _physics_loss_at_target_lambda(
            model,
            test2d,
        ),
        "test_initial_displacement_loss": float(
            common_losses["initial_displacement"].numpy()
        ),
        "test_initial_velocity_loss": float(
            common_losses["initial_velocity"].numpy()
        ),
        "test_boundary_loss": float(common_losses["boundary"].numpy()),
        "lambda_trained": lambda_trained,
        "lambda_1": lambda_value,
        "lambda_error_percent": lambda_error,
        "parameter_count": int(model.count_params()),
        "elapsed_seconds": float(elapsed),
        "convergence": convergence,
        "initialization": initialization,
        "weights": artifacts,
        "training_status": "trained_in_current_run",
    }


def summarize_results_v2(results):
    target_summaries = summarize_grouped_targets(
        results=results,
        group_names=configured_methods,
        thresholds=comparison_thresholds,
        expected_runs=len(configured_seeds),
    )
    primary_threshold = float(comparison_stop_cfg["threshold"])
    summaries = []
    for method_name in configured_methods:
        rows = [row for row in results if row["method"] == method_name]
        if not rows:
            continue

        def stats(key):
            values = [float(row[key]) for row in rows if row.get(key) is not None]
            if not values:
                return None, None
            return (
                float(np.mean(values)),
                float(np.std(values, ddof=1)) if len(values) > 1 else 0.0,
            )

        primary = next(
            item
            for item in target_summaries
            if item["group"] == method_name
            and np.isclose(item["threshold"], primary_threshold)
        )
        l1_mean, l1_std = stats("full_grid_relative_l1")
        l2_mean, l2_std = stats("full_grid_relative_l2")
        physics_mean, physics_std = stats("test_physics_loss_target_lambda")
        lambda_mean, lambda_std = stats("lambda_error_percent")
        time_mean, time_std = stats("elapsed_seconds")
        iteration_mean, iteration_std = stats("iterations_completed")
        summaries.append(
            {
                "method": method_name,
                "completed_seeds": len(rows),
                "expected_seeds": len(configured_seeds),
                "relative_l1_mean": l1_mean,
                "relative_l1_std": l1_std,
                "relative_l2_mean": l2_mean,
                "relative_l2_std": l2_std,
                "target_lambda_physics_loss_mean": physics_mean,
                "target_lambda_physics_loss_std": physics_std,
                "lambda_error_percent_mean": lambda_mean,
                "lambda_error_percent_std": lambda_std,
                "elapsed_seconds_mean": time_mean,
                "elapsed_seconds_std": time_std,
                "iterations_completed_mean": iteration_mean,
                "iterations_completed_std": iteration_std,
                "target_reached_count": primary["reached_runs"],
                "target_available_count": primary["available_runs"],
                "median_target_iteration": primary["median_iteration_reached"],
                "median_target_elapsed_seconds": primary[
                    "median_elapsed_seconds_reached"
                ],
            }
        )
    return summaries, target_summaries


def _format_optional(value, format_spec):
    return "N/A" if value is None else format(value, format_spec)


def render_comparison_document_v2(results, summaries, target_summaries, status):
    target_percent = float(comparison_stop_cfg["threshold"]) * 100.0
    lines = [
        "# 2D Supervised-only and PINN Method Comparison V2",
        "",
        f"Status: {status}",
        f"Protocol version: {method_comparison_cfg['protocol_version']}",
        f"Configuration fingerprint: `{comparison_fingerprint}`",
        "",
        "## 비교 구성",
        "",
        "2D 비교에 포함된 방법과 seed는 다음과 같다.",
        "",
        f"- 방법: {', '.join(configured_methods)}",
        f"- Seed: {', '.join(str(seed) for seed in configured_seeds)}",
        f"- 총 실험 수: {expected_experiment_count}",
        "",
        "## V2 Stopping Protocol",
        "",
        f"- Maximum iterations: {method_comparison_cfg['maximum_iterations']}",
        f"- Evaluation interval: {method_comparison_cfg['evaluation_interval']}",
        f"- Stopping target: Relative L1 < {target_percent:g}% and Relative L2 < {target_percent:g}%",
        "- Evaluation field: fixed analytical full-grid",
        "- A run stops only when both error targets are reached or the maximum iteration cap is reached.",
        "- The exact grid is used as a verification benchmark, not as a production validation set.",
        "- Iteration and wall time are both reported because per-iteration cost differs by method.",
        "- Supervised-only does not train lambda_1.",
        "- No model is promoted automatically.",
        "",
        "## Four-seed Aggregate",
        "",
        "| Method | Seeds | Target reached | Median target iteration | Median target time (s) | Final Relative L2 mean | Target-lambda physics mean |",
        "|---|---:|---:|---:|---:|---:|---:|",
    ]
    for row in summaries:
        lines.append(
            f"| {row['method']} | {row['completed_seeds']}/{row['expected_seeds']} | "
            f"{row['target_reached_count']}/{row['target_available_count']} | "
            f"{_format_optional(row['median_target_iteration'], '.0f')} | "
            f"{_format_optional(row['median_target_elapsed_seconds'], '.1f')} | "
            f"{row['relative_l2_mean']:.6e} | "
            f"{row['target_lambda_physics_loss_mean']:.6e} |"
        )

    lines.extend(
        [
            "",
            "## Per-run Results",
            "",
            "| Method | Seed | Stop reason | Iterations | Relative L1 | Relative L2 | Time (s) | lambda error |",
            "|---|---:|---|---:|---:|---:|---:|---:|",
        ]
    )
    for row in results:
        lambda_text = (
            "N/A"
            if row["lambda_error_percent"] is None
            else f"{row['lambda_error_percent']:.3f}%"
        )
        lines.append(
            f"| {row['method']} | {row['seed']} | {row['stop_reason']} | "
            f"{row['iterations_completed']} | "
            f"{row['full_grid_relative_l1']:.6e} | "
            f"{row['full_grid_relative_l2']:.6e} | "
            f"{row['elapsed_seconds']:.1f} | {lambda_text} |"
        )

    lines.extend(
        [
            "",
            "## Threshold Ladder",
            "",
            "| Method | Threshold | Available | Reached | Median iteration | Median time (s) |",
            "|---|---:|---:|---:|---:|---:|",
        ]
    )
    for row in target_summaries:
        lines.append(
            f"| {row['group']} | {row['threshold'] * 100:g}% | "
            f"{row['available_runs']}/{row['expected_runs']} | "
            f"{row['reached_runs']}/{row['available_runs']} | "
            f"{_format_optional(row['median_iteration_reached'], '.0f')} | "
            f"{_format_optional(row['median_elapsed_seconds_reached'], '.1f')} |"
        )

    lines.extend(
        [
            "",
            "## Interpretation Boundary",
            "",
            "Target-reaching speed is evaluated together with final field error, physics residual, parameter error, and four-seed stability. Runs that do not satisfy both target metrics within the maximum iteration cap remain explicitly classified as maximum_iterations_reached.",
        ]
    )
    return "\n".join(lines) + "\n"


def save_comparison_state_v2(results, status):
    results = sorted(
        results,
        key=lambda row: (
            configured_seeds.index(int(row["seed"])),
            configured_methods.index(row["method"]),
        ),
    )
    summaries, target_summaries = summarize_results_v2(results)
    payload = {
        "protocol_version": 2,
        "status": status,
        "config_fingerprint": comparison_fingerprint,
        "config": method_comparison_cfg,
        "completed_experiments": len(results),
        "expected_experiments": expected_experiment_count,
        "results": results,
        "summaries": summaries,
        "target_summaries": target_summaries,
        "promotion_gate": {
            "automatic_promotion": False,
            "required_status": "complete",
            "required_seed_count": len(configured_seeds),
            "requires_api_interface_regression": True,
            "requires_api_numerical_regression": True,
            "requires_explicit_selection": True,
        },
    }
    comparison_json_path.write_text(
        json.dumps(payload, ensure_ascii=False, indent=2),
        encoding="utf-8",
    )
    if results:
        flat_rows = []
        primary_key = threshold_key(comparison_stop_cfg["threshold"])
        for row in results:
            primary = row["convergence"]["targets"][primary_key]
            flat_rows.append(
                {
                    "method": row["method"],
                    "seed": row["seed"],
                    "stop_reason": row["stop_reason"],
                    "target_reached": row["target_reached"],
                    "target_iteration": primary["iteration"],
                    "target_elapsed_seconds": primary["elapsed_seconds"],
                    "iterations_completed": row["iterations_completed"],
                    "full_grid_relative_l1": row["full_grid_relative_l1"],
                    "full_grid_relative_l2": row["full_grid_relative_l2"],
                    "target_lambda_physics_loss": row[
                        "test_physics_loss_target_lambda"
                    ],
                    "lambda_error_percent": row["lambda_error_percent"],
                    "elapsed_seconds": row["elapsed_seconds"],
                    "initialization_group": row["initialization"]["group"],
                    "initialization_sha256": row["initialization"]["sha256"],
                    "weights_final": row["weights"]["production_candidate"],
                }
            )
        with comparison_csv_path.open("w", newline="", encoding="utf-8") as stream:
            writer = csv.DictWriter(stream, fieldnames=list(flat_rows[0]))
            writer.writeheader()
            writer.writerows(flat_rows)
    document = render_comparison_document_v2(
        results,
        summaries,
        target_summaries,
        status,
    )
    comparison_report_path.write_text(document, encoding="utf-8")
    comparison_corpus_path.write_text(document, encoding="utf-8")
    return summaries, target_summaries


def train_comparison_method_v2(method_name, method_spec, seed):
    comparison_model, initialization = _initialize_comparison_model(
        method_name,
        method_spec,
        seed,
    )
    schedule = tf.keras.optimizers.schedules.ExponentialDecay(
        initial_learning_rate=wave2d_config["training"]["learning_rate"],
        decay_steps=wave2d_config["training"]["decay_steps"],
        decay_rate=wave2d_config["training"]["decay_rate"],
        staircase=True,
    )
    optimizer = tf.keras.optimizers.Adam(learning_rate=schedule)

    @tf.function
    def method_train_step(iteration):
        with tf.GradientTape() as tape:
            if method_spec["supervised_only"]:
                prediction = comparison_model(
                    train2d["X_supervised"],
                    training=True,
                )
                objective = tf.reduce_mean(
                    tf.square(train2d["u_supervised"] - prediction)
                )
            else:
                objective = calculate_method_losses(
                    comparison_model,
                    train2d,
                    method_spec,
                    iteration,
                    training=True,
                )["total"]
        gradients = tape.gradient(
            objective,
            comparison_model.trainable_variables,
        )
        pairs = [
            (gradient, variable)
            for gradient, variable in zip(
                gradients,
                comparison_model.trainable_variables,
            )
            if gradient is not None
        ]
        optimizer.apply_gradients(pairs)
        return objective

    method_dir = comparison_root / method_name / f"seed_{seed}"
    method_dir.mkdir(parents=True, exist_ok=True)
    best_path = method_dir / "best_target.weights.h5"
    last_path = method_dir / "last.weights.h5"
    final_path = method_dir / "final.weights.h5"
    history_path = method_dir / "training_history.csv"
    target_path = method_dir / "convergence_targets.json"

    tracker = ConvergenceTargetTracker(
        thresholds=comparison_thresholds,
        metric_names=tuple(comparison_stop_cfg["metric_names"]),
        evaluation_interval=method_comparison_cfg["evaluation_interval"],
    )
    best_validation = _supervised_validation_mse(comparison_model)
    best_target_score = np.inf
    history_rows = []
    started_at = time.perf_counter()

    def write_history():
        if not history_rows:
            return
        with history_path.open("w", newline="", encoding="utf-8") as stream:
            writer = csv.DictWriter(stream, fieldnames=list(history_rows[0]))
            writer.writeheader()
            writer.writerows(history_rows)

    def record_checkpoint(iteration, train_objective):
        nonlocal best_validation, best_target_score
        current_validation = _supervised_validation_mse(comparison_model)
        common_validation = calculate_losses_2d(
            comparison_model,
            val2d,
            training=False,
        )
        _, field_metrics = _full_grid_metrics(comparison_model)
        elapsed_at_checkpoint = time.perf_counter() - started_at
        newly_reached = tracker.observe(
            iteration=iteration,
            elapsed_seconds=elapsed_at_checkpoint,
            metrics=field_metrics,
        )
        current_target_score = _target_score(field_metrics)
        if current_target_score < best_target_score:
            best_target_score = current_target_score
            comparison_model.save_weights(best_path)
        best_validation = min(best_validation, current_validation)
        row = {
            "iteration": int(iteration),
            "elapsed_seconds": float(elapsed_at_checkpoint),
            "train_objective": (
                None if train_objective is None else float(train_objective.numpy())
            ),
            "supervised_validation_mse": current_validation,
            "target_score_max_relative_l1_l2": current_target_score,
            "full_grid_relative_l1": field_metrics["relative_l1"],
            "full_grid_relative_l2": field_metrics["relative_l2"],
            "fixed_validation_total": float(common_validation["total"].numpy()),
            "fixed_validation_physics": float(common_validation["physics"].numpy()),
            "lambda_1": (
                None
                if method_spec["supervised_only"]
                else float(comparison_model.lambda_1.numpy())
            ),
        }
        history_rows.append(row)
        write_history()
        snapshot = tracker.snapshot(
            iteration_budget=method_comparison_cfg["maximum_iterations"]
        )
        target_path.write_text(
            json.dumps(snapshot, ensure_ascii=False, indent=2),
            encoding="utf-8",
        )
        lambda_text = "N/A" if row["lambda_1"] is None else f"{row['lambda_1']:.6f}"
        reached_text = "-" if not newly_reached else ",".join(newly_reached)
        objective_text = (
            "initial"
            if row["train_objective"] is None
            else f"{row['train_objective']:.4e}"
        )
        print(
            f"{method_name:<24} seed={seed} iter={iteration:6d} "
            f"objective={objective_text} "
            f"rel_l1={row['full_grid_relative_l1']:.4e} "
            f"rel_l2={row['full_grid_relative_l2']:.4e} "
            f"time={elapsed_at_checkpoint:.1f}s "
            f"targets={reached_text} lambda_1={lambda_text}"
        )
        primary_key = threshold_key(comparison_stop_cfg["threshold"])
        return snapshot["targets"][primary_key]["reached"] is True

    completed_iterations = 0
    target_reached = record_checkpoint(iteration=0, train_objective=None)
    if not target_reached:
        for iteration in range(
            1,
            method_comparison_cfg["maximum_iterations"] + 1,
        ):
            objective = method_train_step(tf.constant(iteration, dtype=tf.int32))
            completed_iterations = iteration
            if (
                iteration % method_comparison_cfg["evaluation_interval"] == 0
                or iteration == method_comparison_cfg["maximum_iterations"]
            ):
                target_reached = record_checkpoint(
                    iteration=iteration,
                    train_objective=objective,
                )
                if target_reached:
                    break

    elapsed = time.perf_counter() - started_at
    comparison_model.save_weights(last_path)
    comparison_model.load_weights(best_path)
    comparison_model.save_weights(final_path)
    stop_reason = (
        "relative_l1_l2_target_reached"
        if target_reached
        else "maximum_iterations_reached"
    )
    convergence = tracker.snapshot(
        iteration_budget=method_comparison_cfg["maximum_iterations"]
    )
    artifacts = {
        "best": str(best_path.relative_to(PROJECT_DIR)),
        "last": str(last_path.relative_to(PROJECT_DIR)),
        "production_candidate": str(final_path.relative_to(PROJECT_DIR)),
        "history": str(history_path.relative_to(PROJECT_DIR)),
        "convergence_targets": str(target_path.relative_to(PROJECT_DIR)),
    }
    return evaluate_comparison_model_v2(
        comparison_model,
        method_name,
        method_spec,
        seed,
        best_validation,
        elapsed,
        completed_iterations,
        stop_reason,
        convergence,
        artifacts,
        initialization,
    )


existing_results = {}
if comparison_json_path.exists() and REUSE_COMPLETED_WAVE2D_METHODS:
    existing_payload = json.loads(
        comparison_json_path.read_text(encoding="utf-8")
    )
    if (
        existing_payload.get("protocol_version") == 2
        and existing_payload.get("config_fingerprint") == comparison_fingerprint
    ):
        for row in existing_payload.get("results", []):
            artifacts = row.get("weights", {})
            required = [
                PROJECT_DIR / artifacts[name]
                for name in (
                    "production_candidate",
                    "history",
                    "convergence_targets",
                )
                if artifacts.get(name)
            ]
            if len(required) == 3 and all(path.exists() for path in required):
                existing_results[(row["method"], int(row["seed"]))] = row

comparison_results = list(existing_results.values())
if RUN_WAVE2D_METHOD_COMPARISON:
    max_new_trials = int(MAX_NEW_COMPARISON_TRIALS_PER_RUN)
    if max_new_trials <= 0:
        raise ValueError("MAX_NEW_COMPARISON_TRIALS_PER_RUN must be positive")
    new_trial_count = 0
    for seed in configured_seeds:
        for method_name in configured_methods:
            key = (method_name, seed)
            if key in existing_results:
                print(
                    f"{method_name:<24} seed={seed} reused completed V2 artifact"
                )
                continue
            if new_trial_count >= max_new_trials:
                break
            result = train_comparison_method_v2(
                method_name,
                method_specs[method_name],
                seed,
            )
            comparison_results.append(result)
            existing_results[key] = result
            new_trial_count += 1
            status = (
                "complete"
                if len(comparison_results) == expected_experiment_count
                else "partial"
            )
            save_comparison_state_v2(comparison_results, status)
        if new_trial_count >= max_new_trials:
            break

    final_status = (
        "complete"
        if len(comparison_results) == expected_experiment_count
        else "partial"
    )
    summaries, target_summaries = save_comparison_state_v2(
        comparison_results,
        final_status,
    )
    print(f"V2 comparison status : {final_status}")
    print(f"Completed experiments : {len(comparison_results)}/{expected_experiment_count}")
    print(f"New trials this run   : {new_trial_count}/{max_new_trials}")

    if final_status == "complete":
        labels = [row["method"] for row in summaries]
        relative_l2_means = [row["relative_l2_mean"] * 100.0 for row in summaries]
        relative_l2_stds = [row["relative_l2_std"] * 100.0 for row in summaries]
        target_iterations = [
            np.nan if row["median_target_iteration"] is None
            else row["median_target_iteration"]
            for row in summaries
        ]
        reached_counts = [row["target_reached_count"] for row in summaries]
        figure_dir = PROJECT_DIR / "report_artifacts" / "figures"
        figure_dir.mkdir(parents=True, exist_ok=True)
        comparison_figure_path = figure_dir / "wave2d_method_comparison_v2.png"
        comparison_figure_pdf_path = figure_dir / "wave2d_method_comparison_v2.pdf"
        fig, axes = plt.subplots(2, 1, figsize=(10.0, 8.5), constrained_layout=True)
        axes[0].bar(
            labels,
            relative_l2_means,
            yerr=relative_l2_stds,
            capsize=4,
            color="0.35",
        )
        axes[0].set_ylabel("Final Relative L2 mean ± std (%)")
        axes[0].set_title("Four-seed Final Field Error")
        axes[1].bar(labels, target_iterations, color="0.55")
        axes[1].set_ylabel("Median iteration to L1/L2 target")
        axes[1].set_title("Target-reaching Efficiency")
        for index, count in enumerate(reached_counts):
            axes[1].text(
                index,
                0.0,
                f"{count}/{len(configured_seeds)} reached",
                ha="center",
                va="bottom",
                rotation=90,
            )
        for axis in axes:
            axis.tick_params(axis="x", rotation=20)
            axis.grid(True, axis="y", linestyle="--", linewidth=0.5, alpha=0.35)
        save_publication_figure(
            fig,
            comparison_figure_path,
            comparison_figure_pdf_path,
        )
        plt.show()
        plt.close(fig)

    print(f"Comparison JSON : {comparison_json_path}")
    print(f"Comparison CSV  : {comparison_csv_path}")
    print(f"Comparison RAG  : {comparison_corpus_path}")
    print(f"Initialization  : {initialization_root}")
else:
    print("2D V2 method comparison is disabled by configuration.")
````

### Production Promotion and API Regression Gate

Automatic production promotion is disabled by default, but promotion is not permanently prohibited. A candidate can be selected after all 24 experiments complete and the four-seed aggregate is reviewed.

The gate is required because Fourier and plain models have different input-layer shapes, metric rankings may disagree, and replacing a weight file without its matching model configuration can break inference. Promotion therefore requires an explicit method and seed or an approved aggregate-selection policy, a matching configuration and initialization manifest, checksum validation, model-load verification, and API regression tests.

API regression has two scopes:

1. **Interface regression** verifies that endpoint paths, request and response schemas, status codes, batch ordering, validation behavior, and OpenAPI contracts remain unchanged.
2. **Numerical regression** allows predictions to change after a model upgrade but verifies finite outputs, deterministic repeated inference, single/batch consistency, domain handling, coefficient validity, reference-grid error, and documented tolerances.

The current verified 2D baseline remains active until the candidate passes both scopes. Versioned artifacts and the previous production manifest provide a rollback path.


````python
# Cell 10E - 2D Wave-PINN 추론 모듈, Schema, Service 및 Route 생성

import importlib


wave2d_module_path = PROJECT_DIR / "app" / "ml" / "wave_pinn_2d.py"
wave2d_module_code = r'''
from pathlib import Path
import json
import numpy as np
import tensorflow as tf


class WavePINN2D(tf.keras.Model):
    def __init__(self, config):
        super().__init__()
        model_cfg = config["model"]
        domain_cfg = config["domain"]
        equation_cfg = config["equation"]
        self.lb = tf.constant(
            [domain_cfg["x_min"], domain_cfg["y_min"], domain_cfg["t_min"]],
            dtype=tf.float32,
        )
        self.ub = tf.constant(
            [domain_cfg["x_max"], domain_cfg["y_max"], domain_cfg["t_max"]],
            dtype=tf.float32,
        )
        self.hidden_layers = [
            tf.keras.layers.Dense(
                units,
                activation=model_cfg["activation"],
                kernel_initializer="glorot_normal",
                bias_initializer="zeros",
            )
            for units in model_cfg["layers"][1:-1]
        ]
        self.output_layer = tf.keras.layers.Dense(
            model_cfg["layers"][-1],
            activation=None,
        )
        self.lambda_1 = self.add_weight(
            name="lambda_1",
            shape=(),
            initializer=tf.keras.initializers.Constant(
                equation_cfg["lambda_initial"]
            ),
            trainable=True,
            dtype=tf.float32,
        )

    def call(self, X, training=False):
        X = tf.cast(X, tf.float32)
        hidden = 2.0 * (X - self.lb) / (self.ub - self.lb) - 1.0
        for layer in self.hidden_layers:
            hidden = layer(hidden, training=training)
        return self.output_layer(hidden, training=training)


class WavePINN2DPredictor:
    def __init__(self, config_path, weights_path):
        self.config_path = Path(config_path)
        self.weights_path = Path(weights_path)
        self.config = json.loads(self.config_path.read_text(encoding="utf-8"))
        self.model = WavePINN2D(self.config)
        self.model(tf.zeros((1, 3), dtype=tf.float32))
        self.model.load_weights(self.weights_path)

    @property
    def lambda_1(self):
        return float(self.model.lambda_1.numpy())

    def predict(self, x, y, t):
        coordinates = np.array([[x, y, t]], dtype=np.float32)
        prediction = self.model(coordinates, training=False)
        return float(prediction.numpy()[0, 0])

    def predict_batch(self, points):
        coordinates = np.asarray(points, dtype=np.float32)
        if coordinates.ndim != 2 or coordinates.shape[1] != 3:
            raise ValueError("points must have shape (N, 3)")
        return self.model(coordinates, training=False).numpy().reshape(-1).tolist()

    def model_info(self):
        domain = self.config["domain"]
        equation = self.config["equation"]
        return {
            "equation": equation["name"],
            "lambda_1": self.lambda_1,
            "lambda_target": equation["lambda_target"],
            "wave_speed": equation["wave_speed"],
            "x_range": [domain["x_min"], domain["x_max"]],
            "y_range": [domain["y_min"], domain["y_max"]],
            "t_range": [domain["t_min"], domain["t_max"]],
        }
'''

wave2d_schema_path = PROJECT_DIR / "app" / "schemas" / "wave2d.py"
wave2d_schema_code = r'''
from pydantic import BaseModel, Field, FiniteFloat


class Wave2DPoint(BaseModel):
    x: FiniteFloat = Field(..., ge=-1.0, le=1.0)
    y: FiniteFloat = Field(..., ge=0.0, le=2.0)
    t: FiniteFloat = Field(..., ge=0.0, le=1.0)


class Wave2DPredictionRequest(Wave2DPoint):
    pass


class Wave2DPredictionResponse(BaseModel):
    x: float
    y: float
    t: float
    u: float
    lambda_1: float


class Wave2DBatchPredictionRequest(BaseModel):
    points: list[Wave2DPoint] = Field(..., min_length=1, max_length=1000)


class Wave2DBatchResult(BaseModel):
    x: float
    y: float
    t: float
    u: float


class Wave2DBatchPredictionResponse(BaseModel):
    predictions: list[Wave2DBatchResult]
    lambda_1: float


class Wave2DModelInfoResponse(BaseModel):
    equation: str
    lambda_1: float
    lambda_target: float
    wave_speed: float
    x_range: list[float]
    y_range: list[float]
    t_range: list[float]
'''

wave2d_service_path = PROJECT_DIR / "app" / "services" / "wave2d_service.py"
wave2d_service_code = r'''
from functools import lru_cache
from pathlib import Path

from app.ml.wave_pinn_2d import WavePINN2DPredictor


PROJECT_DIR = Path(__file__).resolve().parents[2]
CONFIG_PATH = PROJECT_DIR / "config" / "wave2d_config.json"
WEIGHTS_PATH = PROJECT_DIR / "model_artifacts" / "wave_pinn_2d" / "final.weights.h5"


class Wave2DService:
    def __init__(self):
        if not CONFIG_PATH.exists():
            raise FileNotFoundError(f"Config file not found: {CONFIG_PATH}")
        if not WEIGHTS_PATH.exists():
            raise FileNotFoundError(f"Model weights not found: {WEIGHTS_PATH}")
        self.predictor = WavePINN2DPredictor(CONFIG_PATH, WEIGHTS_PATH)

    def predict(self, x, y, t):
        return {
            "x": x,
            "y": y,
            "t": t,
            "u": self.predictor.predict(x, y, t),
            "lambda_1": self.predictor.lambda_1,
        }

    def predict_batch(self, points):
        coordinates = [[point.x, point.y, point.t] for point in points]
        predictions = self.predictor.predict_batch(coordinates)
        return {
            "predictions": [
                {
                    "x": point.x,
                    "y": point.y,
                    "t": point.t,
                    "u": prediction,
                }
                for point, prediction in zip(points, predictions)
            ],
            "lambda_1": self.predictor.lambda_1,
        }

    def model_info(self):
        return self.predictor.model_info()


@lru_cache
def get_wave2d_service():
    return Wave2DService()
'''

wave2d_route_path = PROJECT_DIR / "app" / "api" / "routes" / "wave2d.py"
wave2d_route_code = r'''
from fastapi import APIRouter, Depends

from app.schemas.wave2d import (
    Wave2DPredictionRequest,
    Wave2DPredictionResponse,
    Wave2DBatchPredictionRequest,
    Wave2DBatchPredictionResponse,
    Wave2DModelInfoResponse,
)
from app.services.wave2d_service import Wave2DService, get_wave2d_service


router = APIRouter(prefix="/wave2d", tags=["2D Wave-PINN"])


@router.post("/predict", response_model=Wave2DPredictionResponse)
def predict_wave2d(
    request: Wave2DPredictionRequest,
    service: Wave2DService = Depends(get_wave2d_service),
):
    return service.predict(request.x, request.y, request.t)


@router.post("/predict/batch", response_model=Wave2DBatchPredictionResponse)
def predict_wave2d_batch(
    request: Wave2DBatchPredictionRequest,
    service: Wave2DService = Depends(get_wave2d_service),
):
    return service.predict_batch(request.points)


@router.get("/info", response_model=Wave2DModelInfoResponse)
def get_wave2d_info(
    service: Wave2DService = Depends(get_wave2d_service),
):
    return service.model_info()
'''

for path, code in [
    (wave2d_module_path, wave2d_module_code),
    (wave2d_schema_path, wave2d_schema_code),
    (wave2d_service_path, wave2d_service_code),
    (wave2d_route_path, wave2d_route_code),
]:
    path.write_text(code.strip() + "\n", encoding="utf-8")

import app.ml.wave_pinn_2d as wave2d_module
import app.schemas.wave2d as wave2d_schema_module
import app.services.wave2d_service as wave2d_service_module
import app.api.routes.wave2d as wave2d_route_module

wave2d_module = importlib.reload(wave2d_module)
wave2d_schema_module = importlib.reload(wave2d_schema_module)
wave2d_service_module = importlib.reload(wave2d_service_module)
wave2d_route_module = importlib.reload(wave2d_route_module)

print(f"2D ML module : {wave2d_module_path}")
print(f"2D Schema    : {wave2d_schema_path}")
print(f"2D Service   : {wave2d_service_path}")
print(f"2D Route     : {wave2d_route_path}")
````

````python
# Cell 19 - RAG 문서 로딩 및 Chunking 모듈 생성

rag_loader_path = (
    PROJECT_DIR
    / "app"
    / "rag"
    / "document_loader.py"
)

rag_loader_code = '''
from dataclasses import asdict, dataclass
from pathlib import Path
import json

from pypdf import PdfReader


@dataclass
class DocumentChunk:
    chunk_id: int
    source: str
    text: str


def load_document(file_path: Path) -> str:
    suffix = file_path.suffix.lower()

    if suffix in {".md", ".txt"}:
        return file_path.read_text(
            encoding="utf-8"
        )

    if suffix == ".pdf":
        reader = PdfReader(
            str(file_path)
        )

        pages = [
            page.extract_text() or ""
            for page in reader.pages
        ]

        return "\\n\\n".join(
            pages
        )

    raise ValueError(
        f"Unsupported document type: {suffix}"
    )


def split_text(
    text: str,
    max_chars: int = 700,
    overlap_paragraphs: int = 1,
) -> list[str]:
    paragraphs = [
        paragraph.strip()
        for paragraph in text.split("\\n\\n")
        if paragraph.strip()
    ]

    chunks = []
    current = []

    for paragraph in paragraphs:
        candidate = "\\n\\n".join(
            current + [paragraph]
        )

        if (
            current
            and len(candidate) > max_chars
        ):
            chunks.append(
                "\\n\\n".join(current)
            )

            current = (
                current[-overlap_paragraphs:]
                if overlap_paragraphs > 0
                else []
            )

        current.append(
            paragraph
        )

    if current:
        chunks.append(
            "\\n\\n".join(current)
        )

    return chunks


def build_chunks(
    documents_dir: Path,
    max_chars: int = 700,
) -> list[DocumentChunk]:
    supported_extensions = {
        ".md",
        ".txt",
        ".pdf",
    }

    document_paths = sorted(
        path
        for path in documents_dir.iterdir()
        if path.suffix.lower()
        in supported_extensions
    )

    chunks = []
    chunk_id = 0

    for document_path in document_paths:
        text = load_document(
            document_path
        )

        document_chunks = split_text(
            text=text,
            max_chars=max_chars,
        )

        for chunk_text in document_chunks:
            chunks.append(
                DocumentChunk(
                    chunk_id=chunk_id,
                    source=document_path.name,
                    text=chunk_text,
                )
            )

            chunk_id += 1

    return chunks


def save_chunks(
    chunks: list[DocumentChunk],
    output_path: Path,
):
    output_path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    data = [
        asdict(chunk)
        for chunk in chunks
    ]

    with open(
        output_path,
        "w",
        encoding="utf-8",
    ) as file:
        json.dump(
            data,
            file,
            ensure_ascii=False,
            indent=2,
        )
'''


with open(
    rag_loader_path,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        rag_loader_code.strip()
        + "\n"
    )


from app.rag.document_loader import (
    build_chunks,
    save_chunks,
)


chunks = build_chunks(
    documents_dir=documents_dir,
    max_chars=700,
)

chunks_path = (
    PROJECT_DIR
    / "rag_artifacts"
    / "document_chunks.json"
)

save_chunks(
    chunks=chunks,
    output_path=chunks_path,
)


print(
    f"Documents : "
    f"{len(documents)}"
)

print(
    f"Chunks    : "
    f"{len(chunks)}"
)

print(
    f"Artifact  : "
    f"{chunks_path}"
)

print()

for chunk in chunks:
    print(
        f"[{chunk.chunk_id}] "
        f"{chunk.source} "
        f"({len(chunk.text)} chars)"
    )
````

````python
# Cell 20 - Semantic Embedding 및 FAISS Index 구축

semantic_module_path = (
    PROJECT_DIR
    / "app"
    / "rag"
    / "semantic_retriever.py"
)

semantic_module_code = '''
from pathlib import Path
import json

import faiss
import numpy as np
from sentence_transformers import SentenceTransformer


class SemanticRetriever:
    def __init__(
        self,
        model_name: str,
        chunks_path: Path,
        index_path: Path,
    ):
        self.model_name = model_name
        self.chunks_path = Path(chunks_path)
        self.index_path = Path(index_path)

        with open(
            self.chunks_path,
            "r",
            encoding="utf-8",
        ) as file:
            self.chunks = json.load(file)

        self.model = SentenceTransformer(
            self.model_name
        )

        self.index = None

    def build_index(self):
        passages = [
            f"passage: {chunk['text']}"
            for chunk in self.chunks
        ]

        embeddings = self.model.encode(
            passages,
            normalize_embeddings=True,
            convert_to_numpy=True,
            show_progress_bar=False,
        )

        embeddings = np.asarray(
            embeddings,
            dtype=np.float32,
        )

        dimension = embeddings.shape[1]

        self.index = faiss.IndexFlatIP(
            dimension
        )

        self.index.add(
            embeddings
        )

        self.index_path.parent.mkdir(
            parents=True,
            exist_ok=True,
        )

        faiss.write_index(
            self.index,
            str(self.index_path),
        )

        return embeddings.shape

    def load_index(self):
        self.index = faiss.read_index(
            str(self.index_path)
        )

    def search(
        self,
        query: str,
        top_k: int = 3,
    ):
        if self.index is None:
            self.load_index()

        query_embedding = self.model.encode(
            [f"query: {query}"],
            normalize_embeddings=True,
            convert_to_numpy=True,
            show_progress_bar=False,
        )

        query_embedding = np.asarray(
            query_embedding,
            dtype=np.float32,
        )

        top_k = min(
            top_k,
            len(self.chunks),
        )

        scores, indices = self.index.search(
            query_embedding,
            top_k,
        )

        results = []

        for rank, (
            score,
            index,
        ) in enumerate(
            zip(
                scores[0],
                indices[0],
            ),
            start=1,
        ):
            chunk = self.chunks[
                int(index)
            ]

            results.append(
                {
                    "rank": rank,
                    "score": float(score),
                    "chunk_id": chunk["chunk_id"],
                    "source": chunk["source"],
                    "text": chunk["text"],
                }
            )

        return results
'''


with open(
    semantic_module_path,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        semantic_module_code.strip()
        + "\n"
    )


from app.rag.semantic_retriever import (
    SemanticRetriever,
)


embedding_model = (
    "intfloat/multilingual-e5-small"
)

faiss_index_path = (
    PROJECT_DIR
    / "rag_artifacts"
    / "faiss.index"
)


semantic_retriever = SemanticRetriever(
    model_name=embedding_model,
    chunks_path=chunks_path,
    index_path=faiss_index_path,
)

embedding_shape = (
    semantic_retriever.build_index()
)


print(
    f"Embedding model : "
    f"{embedding_model}"
)

print(
    f"Embedding shape : "
    f"{embedding_shape}"
)

print(
    f"FAISS vectors   : "
    f"{semantic_retriever.index.ntotal}"
)

print(
    f"FAISS index     : "
    f"{faiss_index_path}"
)
````

````python
# Cell 21 - BM25 Keyword Retriever 생성

bm25_module_path = (
    PROJECT_DIR
    / "app"
    / "rag"
    / "keyword_retriever.py"
)

bm25_module_code = '''
from pathlib import Path
import json
import re

from rank_bm25 import BM25Okapi


class KeywordRetriever:
    def __init__(
        self,
        chunks_path: Path,
    ):
        self.chunks_path = Path(
            chunks_path
        )

        with open(
            self.chunks_path,
            "r",
            encoding="utf-8",
        ) as file:
            self.chunks = json.load(
                file
            )

        self.tokenized_corpus = [
            self.tokenize(
                chunk["text"]
            )
            for chunk in self.chunks
        ]

        self.bm25 = BM25Okapi(
            self.tokenized_corpus
        )

    @staticmethod
    def tokenize(
        text: str,
    ) -> list[str]:
        return re.findall(
            r"[A-Za-z0-9_./-]+|[가-힣]+",
            text.lower(),
        )

    def search(
        self,
        query: str,
        top_k: int = 3,
    ):
        query_tokens = self.tokenize(
            query
        )

        scores = self.bm25.get_scores(
            query_tokens
        )

        top_k = min(
            top_k,
            len(self.chunks),
        )

        ranked_indices = sorted(
            range(len(scores)),
            key=lambda index: scores[index],
            reverse=True,
        )[:top_k]

        results = []

        for rank, index in enumerate(
            ranked_indices,
            start=1,
        ):
            chunk = self.chunks[
                index
            ]

            results.append(
                {
                    "rank": rank,
                    "score": float(
                        scores[index]
                    ),
                    "chunk_id": chunk[
                        "chunk_id"
                    ],
                    "source": chunk[
                        "source"
                    ],
                    "text": chunk[
                        "text"
                    ],
                }
            )

        return results
'''


with open(
    bm25_module_path,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        bm25_module_code.strip()
        + "\n"
    )


from app.rag.keyword_retriever import (
    KeywordRetriever,
)


keyword_retriever = KeywordRetriever(
    chunks_path=chunks_path
)


print(
    f"BM25 documents : "
    f"{len(keyword_retriever.chunks)}"
)

print(
    f"BM25 module    : "
    f"{bm25_module_path}"
)
````

````python
# Cell 22 - Semantic Search와 BM25를 결합한 Hybrid Retriever 생성

hybrid_module_path = (
    PROJECT_DIR
    / "app"
    / "rag"
    / "hybrid_retriever.py"
)

hybrid_module_code = '''
from pathlib import Path

from app.rag.semantic_retriever import SemanticRetriever
from app.rag.keyword_retriever import KeywordRetriever


class HybridRetriever:
    def __init__(
        self,
        model_name: str,
        chunks_path: Path,
        index_path: Path,
        semantic_weight: float = 0.6,
        keyword_weight: float = 0.4,
        rrf_k: int = 60,
    ):
        self.semantic_weight = semantic_weight
        self.keyword_weight = keyword_weight
        self.rrf_k = rrf_k

        self.semantic_retriever = SemanticRetriever(
            model_name=model_name,
            chunks_path=chunks_path,
            index_path=index_path,
        )

        self.keyword_retriever = KeywordRetriever(
            chunks_path=chunks_path
        )

    def search(
        self,
        query: str,
        top_k: int = 3,
        candidate_k: int = 5,
    ):
        semantic_results = (
            self.semantic_retriever.search(
                query=query,
                top_k=candidate_k,
            )
        )

        keyword_results = (
            self.keyword_retriever.search(
                query=query,
                top_k=candidate_k,
            )
        )

        combined = {}

        for result in semantic_results:
            chunk_id = result["chunk_id"]

            combined[chunk_id] = {
                "chunk_id": chunk_id,
                "source": result["source"],
                "text": result["text"],
                "semantic_rank": result["rank"],
                "semantic_score": result["score"],
                "keyword_rank": None,
                "keyword_score": None,
                "hybrid_score": 0.0,
            }

            combined[chunk_id]["hybrid_score"] += (
                self.semantic_weight
                / (
                    self.rrf_k
                    + result["rank"]
                )
            )

        for result in keyword_results:
            chunk_id = result["chunk_id"]

            if chunk_id not in combined:
                combined[chunk_id] = {
                    "chunk_id": chunk_id,
                    "source": result["source"],
                    "text": result["text"],
                    "semantic_rank": None,
                    "semantic_score": None,
                    "keyword_rank": result["rank"],
                    "keyword_score": result["score"],
                    "hybrid_score": 0.0,
                }

            else:
                combined[chunk_id][
                    "keyword_rank"
                ] = result["rank"]

                combined[chunk_id][
                    "keyword_score"
                ] = result["score"]

            combined[chunk_id]["hybrid_score"] += (
                self.keyword_weight
                / (
                    self.rrf_k
                    + result["rank"]
                )
            )

        ranked_results = sorted(
            combined.values(),
            key=lambda item: item["hybrid_score"],
            reverse=True,
        )

        # Preserve RRF ordering while prioritizing document-level coverage.
        selected_results = []
        selected_sources = set()

        for result in ranked_results:
            source = result["source"]
            if source in selected_sources:
                continue
            selected_results.append(result)
            selected_sources.add(source)
            if len(selected_results) == top_k:
                break

        if len(selected_results) < top_k:
            selected_chunk_ids = {
                result["chunk_id"]
                for result in selected_results
            }
            remaining_results = [
                result
                for result in ranked_results
                if result["chunk_id"] not in selected_chunk_ids
            ]
            selected_results.extend(
                remaining_results[:top_k - len(selected_results)]
            )

        results = []
        for rank, result in enumerate(selected_results, start=1):
            result["rank"] = rank
            results.append(result)

        return results
'''


with open(
    hybrid_module_path,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        hybrid_module_code.strip()
        + "\n"
    )


import importlib
import app.rag.hybrid_retriever as hybrid_retriever_module

hybrid_retriever_module = importlib.reload(hybrid_retriever_module)
HybridRetriever = hybrid_retriever_module.HybridRetriever


hybrid_retriever = HybridRetriever(
    model_name=embedding_model,
    chunks_path=chunks_path,
    index_path=faiss_index_path,
)


print(
    f"Hybrid module : "
    f"{hybrid_module_path}"
)

print(
    f"Semantic weight : "
    f"{hybrid_retriever.semantic_weight}"
)

print(
    f"Keyword weight  : "
    f"{hybrid_retriever.keyword_weight}"
)
````

````python
# Cell 24 - Hybrid Retrieval 기반 RAG Service 생성

rag_service_path = (
    PROJECT_DIR
    / "app"
    / "services"
    / "rag_service.py"
)

rag_service_code = '''
from functools import lru_cache
from pathlib import Path

from app.rag.hybrid_retriever import HybridRetriever


PROJECT_DIR = Path(__file__).resolve().parents[2]

CHUNKS_PATH = (
    PROJECT_DIR
    / "rag_artifacts"
    / "document_chunks.json"
)

INDEX_PATH = (
    PROJECT_DIR
    / "rag_artifacts"
    / "faiss.index"
)

EMBEDDING_MODEL = (
    "intfloat/multilingual-e5-small"
)


class RAGService:
    def __init__(self):
        if not CHUNKS_PATH.exists():
            raise FileNotFoundError(
                f"Chunks file not found: {CHUNKS_PATH}"
            )

        if not INDEX_PATH.exists():
            raise FileNotFoundError(
                f"FAISS index not found: {INDEX_PATH}"
            )

        self.retriever = HybridRetriever(
            model_name=EMBEDDING_MODEL,
            chunks_path=CHUNKS_PATH,
            index_path=INDEX_PATH,
        )

    def search(
        self,
        query: str,
        top_k: int = 3,
    ):
        return self.retriever.search(
            query=query,
            top_k=top_k,
            candidate_k=5,
        )

    def build_context(
        self,
        query: str,
        top_k: int = 3,
    ):
        results = self.search(
            query=query,
            top_k=top_k,
        )

        context_parts = []

        for result in results:
            context_parts.append(
                (
                    f"Source: {result['source']}\\n"
                    f"Chunk ID: {result['chunk_id']}\\n"
                    f"{result['text']}"
                )
            )

        context = "\\n\\n".join(
            context_parts
        )

        return {
            "query": query,
            "context": context,
            "sources": [
                {
                    "source": result["source"],
                    "chunk_id": result["chunk_id"],
                    "score": result["hybrid_score"],
                }
                for result in results
            ],
        }


@lru_cache
def get_rag_service():
    return RAGService()
'''


with open(
    rag_service_path,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        rag_service_code.strip()
        + "\n"
    )


print(
    f"RAG Service : "
    f"{rag_service_path}"
)
````

````python
# Cell 29 - AI Agent Tool 모듈 생성

import importlib


agent_tools_path = PROJECT_DIR / "app" / "agent" / "tools.py"

agent_tools_code = r'''
import ast
import json
import math
import operator
from pathlib import Path


PROJECT_ROOT = Path(__file__).resolve().parents[2]
MAX_EXPRESSION_LENGTH = 200
MAX_AST_NODES = 64


def _finite_number(name, value):
    number = float(value)
    if not math.isfinite(number):
        raise ValueError(f"{name} must be finite")
    return number


def search_project_documents(query: str) -> str:
    "Search the project RAG corpus."
    query = str(query).strip()
    if not query:
        raise ValueError("query must not be blank")
    from app.services.rag_service import get_rag_service

    result = get_rag_service().build_context(query=query, top_k=3)
    return json.dumps(result, ensure_ascii=False)


def predict_wave2d(x: float, y: float, t: float) -> str:
    "Predict the 2D Wave-PINN field."
    x = _finite_number("x", x)
    y = _finite_number("y", y)
    t = _finite_number("t", t)
    if not -1.0 <= x <= 1.0:
        raise ValueError("x must be between -1.0 and 1.0")
    if not 0.0 <= y <= 2.0:
        raise ValueError("y must be between 0.0 and 2.0")
    if not 0.0 <= t <= 1.0:
        raise ValueError("t must be between 0.0 and 1.0")
    from app.services.wave2d_service import get_wave2d_service

    return json.dumps(
        get_wave2d_service().predict(x=x, y=y, t=t),
        ensure_ascii=False,
    )


def _comparison_rows(payload):
    return payload.get("results", []) if isinstance(payload, dict) else []


def compare_training_results(query: str = "") -> str:
    "Return an artifact-backed training-method comparison."
    wave2d_path = (
        PROJECT_ROOT / "report_artifacts" / "wave2d_method_comparison.json"
    )

    if wave2d_path.exists():
        payload = json.loads(wave2d_path.read_text(encoding="utf-8"))
        summaries = payload.get("summaries", [])
        if summaries:
            completed = payload.get("completed_experiments", 0)
            expected = payload.get("expected_experiments", 0)
            lines = [
                f"2D Supervised-only/PINN 비교 상태: "
                f"{payload.get('status', 'unknown')} "
                f"({completed}/{expected} experiments)",
            ]
            for row in summaries:
                lambda_mean = row.get("lambda_error_percent_mean")
                lambda_std = row.get("lambda_error_percent_std")
                lambda_text = (
                    "N/A (Supervised-only)"
                    if lambda_mean is None
                    else f"{lambda_mean:.3f}% ± {lambda_std:.3f}%"
                )
                lines.append(
                    f"- {row['method']}: Relative L2 "
                    f"{row['relative_l2_mean'] * 100.0:.3f}% ± "
                    f"{row['relative_l2_std'] * 100.0:.3f}%, "
                    f"target-lambda physics loss "
                    f"{row['target_lambda_physics_loss_mean']:.6e} ± "
                    f"{row['target_lambda_physics_loss_std']:.6e}, "
                    f"lambda_1 error {lambda_text}, "
                    f"seeds={row['completed_seeds']}/{row['expected_seeds']}"
                )
            if completed == expected and expected > 0:
                field_best = min(
                    summaries,
                    key=lambda row: row["relative_l2_mean"],
                )
                physics_best = min(
                    summaries,
                    key=lambda row: row[
                        "target_lambda_physics_loss_mean"
                    ],
                )
                lines.append(
                    f"4-seed 평균 Field Relative L2 최저: "
                    f"{field_best['method']}"
                )
                lines.append(
                    f"4-seed 평균 target-lambda physics loss 최저: "
                    f"{physics_best['method']}"
                )
            lines.append(
                "Field 오차와 물리 residual은 서로 다른 지표이며, "
                "production model은 자동 승격하지 않는다."
            )
            return "\n".join(lines)

        rows = _comparison_rows(payload)
        if rows:
            lines = [
                f"2D Wave-PINN 기법 비교 상태: {payload.get('status', 'unknown')}",
            ]
            for row in rows:
                physics_loss = row.get(
                    "test_physics_loss_target_lambda",
                    row.get("test_physics_loss"),
                )
                lambda_error = row.get("lambda_error_percent")
                lambda_text = (
                    "N/A"
                    if lambda_error is None
                    else f"{lambda_error:.3f}%"
                )
                lines.append(
                    f"- {row['method']}: Relative L2 "
                    f"{row['full_grid_relative_l2'] * 100.0:.3f}%, "
                    f"physics loss {physics_loss:.6e}, "
                    f"lambda_1 error {lambda_text}"
                )
            if len(rows) == payload.get("expected_experiments"):
                field_best = min(
                    rows,
                    key=lambda row: row["full_grid_relative_l2"],
                )
                physics_best = min(
                    rows,
                    key=lambda row: row.get(
                        "test_physics_loss_target_lambda",
                        row.get("test_physics_loss", float("inf")),
                    ),
                )
                lines.append(
                    f"Field Relative L2 기준 최저 오차: {field_best['method']}"
                )
                lines.append(
                    f"Physics loss 기준 최저 오차: {physics_best['method']}"
                )
            lines.append("완료된 seed 수와 평가 프로토콜을 함께 확인해야 한다.")
            return "\n".join(lines)


    return "비교 결과 artifact가 아직 생성되지 않았다."


_BINARY_OPERATORS = {
    ast.Add: operator.add,
    ast.Sub: operator.sub,
    ast.Mult: operator.mul,
    ast.Div: operator.truediv,
    ast.Pow: operator.pow,
    ast.Mod: operator.mod,
}
_UNARY_OPERATORS = {
    ast.UAdd: operator.pos,
    ast.USub: operator.neg,
}
_ALLOWED_FUNCTIONS = {"abs": abs}


def _evaluate_expression(node):
    if isinstance(node, ast.Expression):
        return _evaluate_expression(node.body)
    if isinstance(node, ast.Constant):
        if isinstance(node.value, bool) or not isinstance(node.value, (int, float)):
            raise ValueError("Only numeric values are allowed")
        return node.value
    if isinstance(node, ast.BinOp):
        operator_type = type(node.op)
        if operator_type not in _BINARY_OPERATORS:
            raise ValueError("Unsupported operator")
        left = _evaluate_expression(node.left)
        right = _evaluate_expression(node.right)
        if operator_type is ast.Pow and abs(right) > 100:
            raise ValueError("Exponent is outside the allowed range")
        return _BINARY_OPERATORS[operator_type](left, right)
    if isinstance(node, ast.UnaryOp):
        operator_type = type(node.op)
        if operator_type not in _UNARY_OPERATORS:
            raise ValueError("Unsupported operator")
        return _UNARY_OPERATORS[operator_type](_evaluate_expression(node.operand))
    if isinstance(node, ast.Call):
        if not isinstance(node.func, ast.Name):
            raise ValueError("Unsupported function")
        function_name = node.func.id
        if function_name not in _ALLOWED_FUNCTIONS or len(node.args) != 1:
            raise ValueError("Unsupported function")
        return _ALLOWED_FUNCTIONS[function_name](_evaluate_expression(node.args[0]))
    raise ValueError("Unsupported expression")


def calculate(expression: str) -> str:
    "Safely calculate a bounded arithmetic expression."
    expression = str(expression).strip()
    if not expression:
        raise ValueError("expression must not be blank")
    if len(expression) > MAX_EXPRESSION_LENGTH:
        raise ValueError("expression is too long")
    tree = ast.parse(expression, mode="eval")
    if sum(1 for _ in ast.walk(tree)) > MAX_AST_NODES:
        raise ValueError("expression is too complex")
    result = _evaluate_expression(tree)
    if isinstance(result, float) and not math.isfinite(result):
        raise ValueError("calculation result must be finite")
    return str(result)


def generate_technical_report() -> str:
    "Generate the artifact-backed technical report."
    from app.reporting.technical_report import TechnicalReportGenerator

    result_path = PROJECT_ROOT / "report_artifacts" / "training_result.json"
    output_path = PROJECT_ROOT / "report_artifacts" / "training_report.md"
    result = TechnicalReportGenerator(result_path=result_path).generate(
        output_path=output_path
    )
    return json.dumps(
        {
            "status": "generated",
            "document": "training_report.md",
            "output_path": result["output_path"],
        },
        ensure_ascii=False,
    )


def generate_feature_catalog() -> str:
    "Generate the current project feature catalog."
    from app.reporting.feature_catalog import FeatureCatalogGenerator

    output_path = PROJECT_ROOT / "report_artifacts" / "feature_catalog.md"
    result = FeatureCatalogGenerator(project_dir=PROJECT_ROOT).generate(
        output_path=output_path
    )
    return json.dumps(
        {
            "status": "generated",
            "document": "feature_catalog.md",
            "output_path": result["output_path"],
            "api_routes": len(result["api_routes"]),
            "agent_tools": len(result["agent_tools"]),
        },
        ensure_ascii=False,
    )
'''

agent_tools_path.write_text(
    agent_tools_code.strip() + "\n",
    encoding="utf-8",
)

import app.agent.tools as tools_module

tools_module = importlib.reload(tools_module)

print(f"Agent tools module : {agent_tools_path}")
for tool_name in [
    "search_project_documents",
    "compare_training_results",
    "predict_wave2d",
    "calculate",
    "generate_technical_report",
    "generate_feature_catalog",
]:
    print(f"- {tool_name}")
````

````python
# Cell 32 - Qwen3.5 Local LLM 기반 Project Agent 생성

import importlib


agent_module_path = PROJECT_DIR / "app" / "agent" / "agent.py"

agent_module_code = r'''
from functools import lru_cache
import json
import re
import uuid

from langchain_core.messages import AIMessage, HumanMessage, SystemMessage, ToolMessage

from app.agent.llm_provider import get_local_chat_model
from app.agent.tools import (
    calculate,
    compare_training_results,
    generate_feature_catalog,
    generate_technical_report,
    predict_wave2d,
    search_project_documents,
)


ROUTER_PROMPT = (
    "You route one project request to exactly one tool.\n\n"
    "Tools:\n"
    "- search_project_documents(query): stored concepts, model results, API usage, architecture, RAG, and evaluation records.\n"
    "- compare_training_results(query): method comparison, comparison superiority, Supervised-only vs PINN, Fourier Feature, Self-Adaptive Loss Weighting, Curriculum, and ablation results.\n"
    "- predict_wave2d(x, y, t): new 2D Wave-PINN inference.\n"
    "- calculate(expression): arithmetic only.\n"
    "- generate_technical_report(): create or regenerate the technical report only when explicitly requested.\n"
    "- generate_feature_catalog(): create or update the feature catalog only when explicitly requested.\n\n"
    "Use the sentinel name no_tool with empty arguments for a greeting that needs no project tool.\n"
    "Existing results are search or comparison requests, not inference requests.\n"
    "The question '학습의 비교우위에 대해 판단해줘' must use compare_training_results.\n"
    "Every field prediction must contain x, y, and t and use predict_wave2d.\n\n"
    "Return only one JSON object:\n"
    "{\"name\": \"tool_name\", \"arguments\": {...}}"
)

ANSWER_PROMPT = (
    "Answer using only the provided tool result. Preserve every numeric value. "
    "Do not invent project values. Answer concisely in the user's language."
)


class PortfolioAgent:
    def __init__(self):
        self.model = get_local_chat_model()
        self.tools = {
            "search_project_documents": search_project_documents,
            "compare_training_results": compare_training_results,
            "predict_wave2d": predict_wave2d,
            "calculate": calculate,
            "generate_technical_report": generate_technical_report,
            "generate_feature_catalog": generate_feature_catalog,
        }

    @staticmethod
    def _message_text(message):
        content = message.content
        if isinstance(content, str):
            return content.strip()
        if isinstance(content, list):
            parts = []
            for item in content:
                if isinstance(item, str):
                    parts.append(item)
                elif isinstance(item, dict):
                    parts.append(str(item.get("text", item)))
                else:
                    parts.append(str(item))
            return "\n".join(parts).strip()
        return str(content).strip()

    @staticmethod
    def _extract_query(messages):
        if not messages:
            raise ValueError("messages must not be empty")
        message = messages[-1]
        if isinstance(message, dict):
            query = message.get("content", "")
        else:
            query = getattr(message, "content", "")
        query = str(query).strip()
        if not query:
            raise ValueError("query must not be blank")
        return query

    @staticmethod
    def _parse_tool_call(text):
        text = text.strip()
        fenced = re.fullmatch(r"```(?:json)?\s*(.*?)\s*```", text, re.DOTALL)
        if fenced:
            text = fenced.group(1).strip()
        tagged = re.search(r"<tool_call>\s*(.*?)\s*</tool_call>", text, re.DOTALL)
        if tagged:
            text = tagged.group(1).strip()

        decoder = json.JSONDecoder()
        parsed = None
        for match in re.finditer(r"\{", text):
            try:
                candidate, _ = decoder.raw_decode(text[match.start():])
            except json.JSONDecodeError:
                continue
            if isinstance(candidate, dict):
                parsed = candidate
                break
        if parsed is None:
            raise ValueError(f"Tool call could not be parsed: {text}")

        name = parsed.get("name") or parsed.get("tool")
        arguments = parsed.get("arguments", parsed.get("args", {}))
        if isinstance(arguments, str):
            arguments = json.loads(arguments)
        if not isinstance(name, str) or not isinstance(arguments, dict):
            raise ValueError("Invalid tool-call structure")
        return {"name": name.strip(), "arguments": arguments}

    def _route(self, query):
        response = self.model.invoke(
            [SystemMessage(content=ROUTER_PROMPT), HumanMessage(content=query)]
        )
        response_text = self._message_text(response)
        try:
            return self._parse_tool_call(response_text)
        except (ValueError, TypeError, json.JSONDecodeError):
            retry = self.model.invoke(
                [
                    SystemMessage(content=ROUTER_PROMPT),
                    HumanMessage(
                        content=(
                            "Return one valid JSON tool call only.\n"
                            f"Original question: {query}\n"
                            f"Invalid output: {response_text}"
                        )
                    ),
                ]
            )
            return self._parse_tool_call(self._message_text(retry))

    @staticmethod
    def _apply_semantic_rules(query, tool_call):
        query_lower = query.lower()
        greeting = query_lower.strip(" .!?")
        if greeting in {"안녕", "안녕하세요", "hello", "hi"}:
            return {"name": "no_tool", "arguments": {}}
        comparison_intent = any(
            term in query_lower
            for term in [
                "비교우위",
                "비교 우위",
                "비교 결과",
                "우세",
                "어떤 기법",
                "method comparison",
                "ablation",
            ]
        )
        comparison_subject = any(
            term in query_lower
            for term in [
                "학습",
                "pinn",
                "supervised",
                "fourier",
                "self-adaptive",
                "self adaptive",
                "curriculum",
            ]
        )
        if comparison_intent and comparison_subject:
            return {
                "name": "compare_training_results",
                "arguments": {"query": query},
            }

        calculator_intent = any(
            term in query_lower
            for term in ["계산", "차이", "calculate", "difference"]
        )
        if calculator_intent:
            number_pattern = (
                r"[-+]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][-+]?\d+)?"
            )
            numbers = re.findall(number_pattern, query)
            expression = ""
            if ("차이" in query or "difference" in query_lower) and len(numbers) >= 2:
                expression = f"abs({numbers[0]} - {numbers[1]})"
            else:
                candidates = [
                    candidate.strip()
                    for candidate in re.findall(
                        r"[0-9eE.()+\-*/%\s]+",
                        query,
                    )
                    if re.search(r"\d", candidate)
                ]
                if candidates:
                    expression = max(
                        candidates,
                        key=lambda candidate: (
                            sum(operator in candidate for operator in "+-*/%"),
                            len(candidate),
                        ),
                    )
            if expression:
                return {
                    "name": "calculate",
                    "arguments": {"expression": expression},
                }

        prediction_intent = any(
            term in query_lower
            for term in ["예측", "predict", "inference"]
        )
        coordinate_pattern = (
            r"[-+]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][-+]?\d+)?"
        )
        coordinates = {}
        for axis in ("x", "y", "t"):
            match = re.search(
                rf"(?<![A-Za-z0-9_]){axis}\s*=\s*({coordinate_pattern})",
                query_lower,
            )
            if match:
                coordinates[axis] = float(match.group(1))
        if prediction_intent and len(coordinates) == 3:
            return {
                "name": "predict_wave2d",
                "arguments": coordinates,
            }

        document_subject = any(
            term in query_lower
            for term in [
                "lambda_1",
                "relative l2",
                "pde",
                "wave speed",
                "파동 속도",
                "hybrid rag",
                "architecture",
                "구조",
                "api",
                "endpoint",
                "저장된",
            ]
        )
        if document_subject:
            return {
                "name": "search_project_documents",
                "arguments": {"query": query},
            }

        report_action = any(
            term in query_lower
            for term in ["생성", "다시", "갱신", "작성", "generate", "regenerate", "update"]
        )
        if report_action and any(
            term in query_lower for term in ["기술보고서", "technical report"]
        ):
            return {"name": "generate_technical_report", "arguments": {}}
        if report_action and any(
            term in query_lower for term in ["기능 문서", "feature catalog"]
        ):
            return {"name": "generate_feature_catalog", "arguments": {}}

        if tool_call.get("name") == "compare_training_results":
            tool_call["arguments"] = {"query": query}

        if tool_call.get("name") == "calculate" and (
            "차이" in query or "difference" in query_lower
        ):
            arguments = tool_call.setdefault("arguments", {})
            expression = str(arguments.get("expression", "")).strip()
            if expression and not expression.startswith("abs("):
                arguments["expression"] = f"abs({expression})"
        return tool_call

    def _execute_tool(self, tool_call):
        tool_name = tool_call.get("name")
        arguments = tool_call.get("arguments", {})
        if tool_name not in self.tools:
            raise ValueError(f"Unknown tool: {tool_name}")
        if not isinstance(arguments, dict):
            raise ValueError("Tool arguments must be an object")
        result = self.tools[tool_name](**arguments)
        return tool_name, arguments, result

    def _generate_answer(self, query, tool_name, tool_result):
        if tool_name in {
            "calculate",
            "compare_training_results",
            "generate_technical_report",
            "generate_feature_catalog",
        }:
            return str(tool_result)
        response = self.model.invoke(
            [
                SystemMessage(content=ANSWER_PROMPT),
                HumanMessage(
                    content=(
                        f"User question:\n{query}\n\n"
                        f"Tool used:\n{tool_name}\n\n"
                        f"Tool result:\n{tool_result}"
                    )
                ),
            ]
        )
        return self._message_text(response)

    def invoke(self, inputs):
        messages = inputs.get("messages", [])
        query = self._extract_query(messages)
        deterministic = self._apply_semantic_rules(
            query,
            {"name": None, "arguments": {}},
        )
        if deterministic.get("name") is not None:
            tool_call = deterministic
        else:
            tool_call = self._apply_semantic_rules(query, self._route(query))
        if tool_call.get("name") == "no_tool":
            greeting_answer = (
                "안녕하세요. 2D Wave-PINN 비교, API, RAG, Agent 구현에 대해 질문할 수 있습니다."
                if any("가" <= character <= "힣" for character in query)
                else "Hello. You can ask about the 2D Wave-PINN comparison, API, RAG, or Agent implementation."
            )
            return {"messages": [*messages, AIMessage(content=greeting_answer)]}
        tool_name, arguments, tool_result = self._execute_tool(tool_call)
        tool_call_id = "tool_" + uuid.uuid4().hex[:12]
        routing_message = AIMessage(
            content="",
            tool_calls=[
                {
                    "name": tool_name,
                    "args": arguments,
                    "id": tool_call_id,
                    "type": "tool_call",
                }
            ],
        )
        tool_message = ToolMessage(
            content=str(tool_result),
            tool_call_id=tool_call_id,
            name=tool_name,
        )
        final_message = AIMessage(
            content=self._generate_answer(query, tool_name, tool_result)
        )
        return {
            "messages": [
                *messages,
                routing_message,
                tool_message,
                final_message,
            ]
        }


@lru_cache
def get_agent():
    return PortfolioAgent()
'''

agent_module_path.write_text(
    agent_module_code.strip() + "\n",
    encoding="utf-8",
)

import app.agent.agent as agent_module

agent_module = importlib.reload(agent_module)
portfolio_agent = agent_module.get_agent()

print(f"Agent module : {agent_module_path}")
print(f"Agent tools  : {len(portfolio_agent.tools)}")
for tool_name in portfolio_agent.tools:
    print(f"- {tool_name}")
````

````python
# Cell 36 - 프로젝트 구현 기능 Feature Catalog 자동 생성

feature_catalog_path = (
    PROJECT_DIR
    / "app"
    / "reporting"
    / "feature_catalog.py"
)

feature_catalog_code = '''
import ast
from pathlib import Path


class FeatureCatalogGenerator:
    def __init__(
        self,
        project_dir,
    ):
        self.project_dir = Path(
            project_dir
        )

    def _get_api_routes(
        self,
    ):
        from app.main import app

        openapi_schema = app.openapi()

        routes = []

        allowed_methods = {
            "get",
            "post",
            "put",
            "patch",
            "delete",
        }

        for path, operations in (
            openapi_schema
            .get(
                "paths",
                {},
            )
            .items()
        ):
            for method in operations:
                if method.lower() not in allowed_methods:
                    continue

                routes.append(
                    {
                        "path": path,
                        "method": method.upper(),
                    }
                )

        return routes

    def _get_agent_tools(
        self,
    ):
        tools_path = (
            self.project_dir
            / "app"
            / "agent"
            / "tools.py"
        )

        tree = ast.parse(
            tools_path.read_text(
                encoding="utf-8"
            )
        )

        target_tools = {
            "search_project_documents",
            "compare_training_results",
            "predict_wave2d",
            "calculate",
            "generate_technical_report",
            "generate_feature_catalog",
        }

        tool_names = []

        for node in tree.body:
            if (
                isinstance(
                    node,
                    (
                        ast.FunctionDef,
                        ast.AsyncFunctionDef,
                    ),
                )
                and node.name
                in target_tools
            ):
                tool_names.append(
                    node.name
                )

        return tool_names

    def _get_rag_components(
        self,
    ):
        rag_dir = (
            self.project_dir
            / "app"
            / "rag"
        )

        component_map = {
            "document_loader.py":
                "Document loading and chunking",

            "semantic_retriever.py":
                "SentenceTransformer embedding and FAISS semantic retrieval",

            "keyword_retriever.py":
                "BM25 keyword retrieval",

            "hybrid_retriever.py":
                "Hybrid retrieval with RRF ranking",
        }

        components = []

        for filename, description in (
            component_map.items()
        ):
            if (
                rag_dir
                / filename
            ).exists():
                components.append(
                    description
                )

        return components

    def _get_reporting_components(
        self,
    ):
        reporting_dir = (
            self.project_dir
            / "app"
            / "reporting"
        )

        components = []

        if (
            reporting_dir
            / "technical_report.py"
        ).exists():
            components.append(
                "Structured training result based Technical Report Generator"
            )

        if (
            reporting_dir
            / "feature_catalog.py"
        ).exists():
            components.append(
                "Automatic Feature Catalog Generator"
            )

        return components

    def generate(
        self,
        output_path,
    ):
        api_routes = (
            self._get_api_routes()
        )

        agent_tools = (
            self._get_agent_tools()
        )

        rag_components = (
            self._get_rag_components()
        )

        reporting_components = (
            self._get_reporting_components()
        )

        api_lines = [
            (
                f'- {route["method"]} '
                f'{route["path"]}'
            )
            for route in api_routes
        ]

        tool_description = {
            "search_project_documents":
                "RAG corpus에서 프로젝트 문서를 검색한다.",

            "predict_wave2d":
                "학습된 2D Wave-PINN을 이용하여 새로운 x, y, t 좌표의 u를 예측한다.",

            "calculate":
                "안전한 arithmetic expression 계산을 수행한다.",

            "generate_technical_report":
                "구조화된 학습 결과에서 기술보고서를 자동 생성한다.",

            "generate_feature_catalog":
                "현재 프로젝트 구현 기능 문서를 자동 생성한다.",
        }

        tool_lines = [
            (
                f"- {name}: "
                f"{tool_description.get(name, '')}"
            )
            for name in agent_tools
        ]

        rag_lines = [
            f"- {component}"
            for component in rag_components
        ]

        reporting_lines = [
            f"- {component}"
            for component in reporting_components
        ]

        report = f"""# AI Engineering Portfolio Feature Catalog

## Project Overview

이 프로젝트는 Wave-PINN 모델을 중심으로
REST API, RAG, AI Agent 및 자동 문서 생성 기능을
하나의 AI engineering workflow로 구성한다.

feature_catalog_type: automatic
feature_source: current project implementation


## Wave-PINN

2D Wave Equation을 주요 대상으로
Physics-Informed Neural Network 기법을 비교한다.


2D Wave-PINN은 공간 좌표 x, y와 시간 좌표 t를 입력받아
wave field u를 예측한다.

PDE coefficient lambda_1도 neural network와 함께 학습한다.

implemented_features:

- 2D wave field reconstruction
- PDE coefficient identification
- Fourier Feature Mapping comparison
- Self-Adaptive Loss Weighting comparison
- Single point inference
- Batch inference
- Input domain validation
- Model artifact loading and reuse


## REST API

FastAPI를 이용하여 학습된 모델과
RAG 기능을 REST endpoint 형태로 제공한다.

registered_endpoints:

{chr(10).join(api_lines)}


## RAG

프로젝트 기술문서를 검색하기 위한
Retrieval-Augmented Generation 구조를 구현한다.

implemented_components:

{chr(10).join(rag_lines)}

retrieval_pipeline:

Document
-> Chunk
-> Multilingual Embedding
-> FAISS Semantic Retrieval
-> BM25 Keyword Retrieval
-> Reciprocal Rank Fusion
-> Hybrid Retrieval Context


## AI Agent

Local Qwen3.5-0.8B 모델을 이용하여
사용자 요청에 적합한 Tool을 선택한다.

implemented_tools:

{chr(10).join(tool_lines)}

agent_flow:

User Query
-> Qwen Tool Routing
-> Tool Execution
-> Tool Result
-> Final Response


## Automated Reporting

학습 및 평가 결과를 구조화된 데이터로 저장하고
이를 이용하여 기술문서를 자동 생성한다.

implemented_components:

{chr(10).join(reporting_lines)}

report_pipeline:

Training Result
-> Structured JSON
-> Technical Report Generator
-> Markdown Report


## Generated Artifacts

현재 프로젝트에서 생성되는 주요 artifact는 다음과 같다.

- model_artifacts/wave_pinn_2d/final.weights.h5
- report_artifacts/training_result.json
- report_artifacts/training_report.md
- report_artifacts/feature_catalog.md
- rag_artifacts/faiss.index
- rag_artifacts/document_chunks.json
- rag_artifacts/retrieval_evaluation.json
- rag_artifacts/agent_tool_evaluation.json


## Architecture Summary

전체 서비스 구조는 다음과 같다.

Request
-> FastAPI Schema
-> Route
-> Service
-> Wave-PINN / RAG / AI Agent
-> Response

문서 생성 workflow는 다음과 같다.

Training
-> Structured Result
-> Technical Report
-> Feature Catalog
-> RAG Corpus
-> Agent Search
"""

        output_path = Path(
            output_path
        )

        output_path.parent.mkdir(
            parents=True,
            exist_ok=True,
        )

        output_path.write_text(
            report,
            encoding="utf-8",
        )

        return {
            "output_path":
                str(
                    output_path
                ),

            "api_routes":
                api_routes,

            "agent_tools":
                agent_tools,

            "rag_components":
                rag_components,

            "reporting_components":
                reporting_components,
        }
'''


feature_catalog_path.write_text(
    feature_catalog_code.strip()
    + "\n",
    encoding="utf-8",
)


import importlib
import app.reporting.feature_catalog as feature_module


feature_module = importlib.reload(
    feature_module
)


feature_output = (
    PROJECT_DIR
    / "report_artifacts"
    / "feature_catalog.md"
)


feature_generator = (
    feature_module
    .FeatureCatalogGenerator(
        project_dir=PROJECT_DIR
    )
)


feature_result = (
    feature_generator.generate(
        output_path=feature_output
    )
)


print(
    f"Feature catalog : "
    f"{feature_result['output_path']}"
)

print(
    f"API routes      : "
    f"{len(feature_result['api_routes'])}"
)

print(
    f"Agent tools     : "
    f"{len(feature_result['agent_tools'])}"
)

print(
    f"RAG components  : "
    f"{len(feature_result['rag_components'])}"
)

print(
    f"Report modules  : "
    f"{len(feature_result['reporting_components'])}"
)

print(
    "Catalog status  : generated"
)
````

````python
# Cell 38 - 공통 RAG Corpus 동기화 및 Index 재생성

import importlib


rag_indexing_path = PROJECT_DIR / "app" / "rag" / "indexing.py"

rag_indexing_code = r'''
import json
import shutil
from pathlib import Path


def read_document(path):
    suffix = path.suffix.lower()
    if suffix in {".md", ".txt"}:
        return path.read_text(encoding="utf-8")
    if suffix == ".pdf":
        from pypdf import PdfReader

        pages = []
        for page in PdfReader(str(path)).pages:
            text = page.extract_text() or ""
            if text.strip():
                pages.append(text.strip())
        return "\n\n".join(pages)
    return ""


def split_document(text, max_chars=700):
    paragraphs = [
        paragraph.strip()
        for paragraph in text.split("\n\n")
        if paragraph.strip()
    ]
    chunks = []
    current = []
    current_length = 0
    for paragraph in paragraphs:
        extra = len(paragraph) + (2 if current else 0)
        if current and current_length + extra > max_chars:
            chunks.append("\n\n".join(current))
            current = [current[-1]]
            current_length = len(current[0])
        current.append(paragraph)
        current_length += len(paragraph) + (2 if len(current) > 1 else 0)
    if current:
        chunks.append("\n\n".join(current))
    return chunks


def synchronize_and_rebuild(
    project_dir,
    generated_documents,
    metadata_filename,
    status,
):
    import faiss
    from sentence_transformers import SentenceTransformer

    project_dir = Path(project_dir)
    documents_dir = project_dir / "data" / "documents"
    reports_dir = project_dir / "report_artifacts"
    artifacts_dir = project_dir / "rag_artifacts"
    documents_dir.mkdir(parents=True, exist_ok=True)
    artifacts_dir.mkdir(parents=True, exist_ok=True)

    for filename in generated_documents:
        source = reports_dir / filename
        if not source.exists():
            raise FileNotFoundError(f"Generated document not found: {source}")
        shutil.copy2(source, documents_dir / filename)

    document_paths = sorted(
        [
            path
            for path in documents_dir.iterdir()
            if path.is_file() and path.suffix.lower() in {".md", ".txt", ".pdf"}
        ],
        key=lambda path: path.name.lower(),
    )
    chunks = []
    for document_path in document_paths:
        for chunk_text in split_document(read_document(document_path)):
            chunks.append(
                {
                    "source": document_path.name,
                    "chunk_id": len(chunks),
                    "text": chunk_text,
                }
            )
    if not chunks:
        raise RuntimeError("RAG corpus produced no chunks")

    chunks_path = artifacts_dir / "document_chunks.json"
    chunks_path.write_text(
        json.dumps(chunks, ensure_ascii=False, indent=2),
        encoding="utf-8",
    )

    embedding_model_name = "intfloat/multilingual-e5-small"
    embedding_model = SentenceTransformer(embedding_model_name)
    embeddings = embedding_model.encode(
        ["passage: " + chunk["text"] for chunk in chunks],
        convert_to_numpy=True,
        normalize_embeddings=True,
        show_progress_bar=True,
    )
    index = faiss.IndexFlatIP(embeddings.shape[1])
    index.add(embeddings.astype("float32"))
    index_path = artifacts_dir / "faiss.index"
    faiss.write_index(index, str(index_path))

    metadata = {
        "status": status,
        "embedding_model": embedding_model_name,
        "documents": [path.name for path in document_paths],
        "document_count": len(document_paths),
        "chunk_count": len(chunks),
        "embedding_dimension": int(embeddings.shape[1]),
        "faiss_vectors": int(index.ntotal),
    }
    metadata_path = artifacts_dir / metadata_filename
    metadata_path.write_text(
        json.dumps(metadata, ensure_ascii=False, indent=2),
        encoding="utf-8",
    )
    return {
        "metadata": metadata,
        "metadata_path": metadata_path,
        "chunks_path": chunks_path,
        "index_path": index_path,
    }
'''

rag_indexing_path.write_text(
    rag_indexing_code.strip() + "\n",
    encoding="utf-8",
)

from app.reporting.feature_catalog import FeatureCatalogGenerator

FeatureCatalogGenerator(project_dir=PROJECT_DIR).generate(
    output_path=PROJECT_DIR / "report_artifacts" / "feature_catalog.md"
)

import app.rag.indexing as indexing_module

indexing_module = importlib.reload(indexing_module)
index_result = indexing_module.synchronize_and_rebuild(
    project_dir=PROJECT_DIR,
    generated_documents=["training_report.md", "feature_catalog.md"],
    metadata_filename="index_metadata.json",
    status="generated_documents_synchronized",
)

import app.services.rag_service as rag_service_module

rag_service_module = importlib.reload(rag_service_module)
if hasattr(rag_service_module.get_rag_service, "cache_clear"):
    rag_service_module.get_rag_service.cache_clear()

metadata = index_result["metadata"]
required_sources = {
    "concepts.md",
    "results.md",
    "guide.md",
    "wave2d.md",
    "training_report.md",
    "feature_catalog.md",
    "wave2d_method_comparison.md",
}
missing_sources = required_sources - set(metadata["documents"])
if missing_sources:
    raise RuntimeError(f"Missing RAG documents: {sorted(missing_sources)}")

print("RAG Corpus Documents")
for document in metadata["documents"]:
    print(f"- {document}")
print(f"\nDocuments       : {metadata['document_count']}")
print(f"Chunks          : {metadata['chunk_count']}")
print(f"FAISS vectors   : {metadata['faiss_vectors']}")
print(f"Chunk artifact  : {index_result['chunks_path']}")
print(f"FAISS artifact  : {index_result['index_path']}")
print("RAG cache       : cleared")
````

````python
# Cell 39 - Agent Schema 및 Service 생성

from pathlib import Path
import importlib


# --------------------------------------------------
# Agent Schema
# --------------------------------------------------

agent_schema_path = (
    PROJECT_DIR
    / "app"
    / "schemas"
    / "agent.py"
)

agent_schema_code = '''
from pydantic import BaseModel, Field, field_validator


class AgentRequest(BaseModel):
    query: str = Field(
        ...,
        min_length=2,
        max_length=500,
        description="Natural language request for the AI Agent",
    )

    @field_validator("query")
    @classmethod
    def validate_query(cls, value):
        value = value.strip()
        if len(value) < 2:
            raise ValueError("query must contain at least two non-space characters")
        return value


class AgentResponse(BaseModel):
    query: str
    answer: str
    tools_used: list[str]
'''


agent_schema_path.write_text(
    agent_schema_code.strip()
    + "\n",
    encoding="utf-8",
)


# --------------------------------------------------
# Agent Service
# --------------------------------------------------

agent_service_path = (
    PROJECT_DIR
    / "app"
    / "services"
    / "agent_service.py"
)

agent_service_code = '''
from functools import lru_cache

from langchain_core.messages import (
    ToolMessage,
)

from app.agent.agent import (
    get_agent,
)


class AgentService:
    def __init__(
        self,
    ):
        self.agent = (
            get_agent()
        )

    def query(
        self,
        query: str,
    ):
        result = (
            self.agent.invoke(
                {
                    "messages": [
                        {
                            "role": "user",
                            "content": query,
                        }
                    ]
                }
            )
        )

        messages = (
            result[
                "messages"
            ]
        )

        tools_used = []

        for message in messages:
            if isinstance(
                message,
                ToolMessage,
            ):
                if (
                    message.name
                    and message.name
                    not in tools_used
                ):
                    tools_used.append(
                        message.name
                    )

        final_message = (
            messages[-1]
        )

        answer = getattr(
            final_message,
            "content",
            str(
                final_message
            ),
        )

        return {
            "query":
                query,

            "answer":
                str(
                    answer
                ),

            "tools_used":
                tools_used,
        }


@lru_cache
def get_agent_service():
    return AgentService()
'''


agent_service_path.write_text(
    agent_service_code.strip()
    + "\n",
    encoding="utf-8",
)


# --------------------------------------------------
# Reload
# --------------------------------------------------

import app.schemas.agent as agent_schema_module
import app.services.agent_service as agent_service_module


agent_schema_module = (
    importlib.reload(
        agent_schema_module
    )
)

agent_service_module = (
    importlib.reload(
        agent_service_module
    )
)


agent_service = (
    agent_service_module
    .get_agent_service()
)


print(
    f"Agent schema  : "
    f"{agent_schema_path}"
)

print(
    f"Agent service : "
    f"{agent_service_path}"
)

print(
    "Request model : "
    "AgentRequest"
)

print(
    "Response model: "
    "AgentResponse"
)

print(
    "Service cache : enabled"
)

print(
    "Agent status  : service ready"
)
````

## Boundary Validation and Calculator Regression Tests

These tests verify finite numeric inputs, whitespace-only queries, batch error locations, calculator allowlist boundaries, division by zero, empty expressions, and expression-size limits without changing the public API endpoint paths or response models.


````python
# Cell 42A - 2D Schema 경계값 및 Calculator 회귀 테스트

import importlib
import json
import math
import sys
from fastapi.testclient import TestClient
from pydantic import ValidationError
from app.agent.tools import calculate
from app.schemas.agent import AgentRequest
from app.schemas.rag import RAGSearchRequest
from app.schemas.wave2d import Wave2DPredictionRequest

validation_results = []
def record_validation(name, passed, detail):
    validation_results.append({"name": name, "passed": bool(passed), "detail": detail})
    print(f"{name:<36}: {'PASS' if passed else 'FAIL'} | {detail}")

for expression, expected in [("(3 + 5) * 2", 16.0), ("abs(0.972986 - 1.0)", 0.027014)]:
    actual_text = calculate(expression)
    actual = float(actual_text)
    passed = math.isclose(actual, expected, rel_tol=0.0, abs_tol=1e-12)
    record_validation(
        f"calculator valid: {expression}",
        passed,
        f"actual={actual_text}, expected={expected}",
    )

for name, expression, expected_error in [
    ("calculator forbidden AST", "__import__('os')", ValueError),
    ("calculator zero division", "1 / 0", ZeroDivisionError),
    ("calculator blank", "   ", ValueError),
    ("calculator exponent limit", "2 ** 101", ValueError),
    ("calculator length limit", "1+" * 101 + "1", ValueError),
]:
    try:
        calculate(expression)
        passed, detail = False, "no exception"
    except expected_error as error:
        passed, detail = True, str(error)
    record_validation(name, passed, detail)

for name, factory in [
    ("Wave2D NaN", lambda: Wave2DPredictionRequest(x=float("nan"), y=1.0, t=0.5)),
    ("Wave2D Infinity", lambda: Wave2DPredictionRequest(x=0.0, y=float("inf"), t=0.5)),
    ("RAG whitespace query", lambda: RAGSearchRequest(query="   ")),
    ("Agent whitespace query", lambda: AgentRequest(query="   ")),
]:
    try:
        factory()
        passed, detail = False, "validation accepted invalid input"
    except ValidationError as error:
        passed, detail = True, str(error.errors()[0].get("loc"))
    record_validation(name, passed, detail)

sys.modules.pop("app.main", None)
importlib.invalidate_caches()
main_module = importlib.import_module("app.main")
client = TestClient(main_module.app)
api_cases = [
    ("Wave2D API NaN", "/wave2d/predict", {"x": "NaN", "y": 1.0, "t": 0.5}),
    ("Wave2D API Infinity", "/wave2d/predict", {"x": 0.0, "y": "Infinity", "t": 0.5}),
    ("Wave2D batch error location", "/wave2d/predict/batch", {"points": [{"x": 2.0, "y": 1.0, "t": 0.5}]}),
    ("RAG API whitespace", "/rag/search", {"query": "   ", "top_k": 3}),
    ("Agent API whitespace", "/agent/query", {"query": "   "}),
]
for name, endpoint, payload in api_cases:
    response = client.post(endpoint, json=payload)
    detail = response.json().get("detail", [])
    record_validation(
        name,
        response.status_code == 422,
        f"status={response.status_code}, detail={str(detail)[:260]}",
    )

validation_path = PROJECT_DIR / "report_artifacts" / "validation_regression_evaluation.json"
validation_path.write_text(json.dumps(validation_results, ensure_ascii=False, indent=2), encoding="utf-8")
passed_count = sum(result["passed"] for result in validation_results)
print(f"Validation regression : {passed_count}/{len(validation_results)}")
if passed_count != len(validation_results):
    raise RuntimeError("Schema or calculator regression test failed")
````
